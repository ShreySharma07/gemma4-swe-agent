### `evalkit/config.py`
Run settings and fixed paths. Stdlib only, so the env step can import it outside the venv.

````python
%%writefile /kaggle/working/evalkit/config.py
"""Run settings and fixed paths. Stdlib only, so the env step can import it outside the venv."""
import json
import pathlib
import time
from dataclasses import asdict, dataclass, field

WORK = pathlib.Path("/kaggle/working")
INPUT = pathlib.Path("/kaggle/input")
VENV = pathlib.Path("/tmp/scorer_host")
TARGET_MODEL = "gemma-4-31b-it-qat-w4a16-ct"
CFG_FILE = WORK / "eval_cfg.json"
SUBMISSION_ZIP = WORK / "submission_src.zip"
ANALYZER = WORK / "trace_analyzer.py"
SANDBOX_ENV = WORK / "sandbox_env.json"   # task-sandbox interpreter + site, written by envsetup
SCORER_SITE = pathlib.Path("/tmp/scorer_site")
SANDBOX_PIP = pathlib.Path("/tmp/sandbox_pip")
VALID_TASKS = WORK / "valid_tasks.json"       # from the preflight sweep; embedded by make_eval_notebook.py


@dataclass
class Settings:
    name: str = "v7_noedit"           # submissions/<name>; label of every output
    task_ids: list | None = None      # None -> all tasks, interleaved by repo (any prefix is a stratified sample)
    max_tasks: int | None = None      # cap the task list (shakedown runs)
    seed: int = 0
    chunk_size: int = 8
    chunk_timeout_min: float = 60     # floor; the runner raises it to chunk_size x (task max_time + 3 min) (Oct 19)
    session_budget_h: float = 10.5    # stop scheduling chunks after this; the session cap is 12 h
    est_task_min: float = 5.5         # 4.5 min agent budget + setup/verification
    concurrency: int = 1              # tasks in flight per Evaluator call (1 = scorer; >1 only for data generation)
    repeats: int = 1                  # >1: run the list again (rep-major) -> pass@k, trajectories for rejection sampling
    devset: bool = False              # True -> renamed dev set (kernel output gemma4-devset-build)
    valid_only: bool = False          # True -> only tasks in valid_tasks.json (gold passes, empty fails; preflight sweep)
    sweep: bool = False               # preflight: gold + empty over ALL tasks -> valid_tasks.json (else 7-task smoke)
    gate_adapters: list | None = None  # behaviour gate: adapter names to compare with base (evalkit/gate.py)
    gate_samples: int = 4             # behaviour gate: samples per decision point
    t0: float = field(default_factory=time.time)  # notebook start; budget counts from here across processes

    @property
    def label(self):
        return f"eval_{self.name}" + ("_dev" if self.devset else "")

    def hours_left(self):
        return self.session_budget_h - (time.time() - self.t0) / 3600

    def save(self, path=CFG_FILE):
        path.write_text(json.dumps(asdict(self), indent=1))

    @classmethod
    def load(cls, path=CFG_FILE):
        return cls(**json.loads(path.read_text()))
````

### `evalkit/util.py`
Logger that writes to stdout and /kaggle/working/run.log, plus a logged shell helper. Stdlib only.

````python
%%writefile /kaggle/working/evalkit/util.py
"""Logger that writes to stdout and /kaggle/working/run.log, plus a logged shell helper. Stdlib only."""
import glob
import subprocess
import time

from .config import WORK


class Log:
    def __init__(self, path=WORK / "run.log"):
        self.f = open(path, "a", buffering=1)

    def __call__(self, *a):
        s = time.strftime("%H:%M:%S ") + " ".join(str(x) for x in a)
        print(s, flush=True)
        self.f.write(s + "\n")

    def sh(self, cmd, timeout=None, check=False):
        self("$", cmd[:220])
        r = subprocess.run(cmd, shell=True, capture_output=True, text=True, timeout=timeout)
        out = ((r.stdout or "")[-1500:] + ("\n" + r.stderr[-1500:] if (r.stderr or "").strip() else "")).strip()
        if out:
            self(out)
        if check and r.returncode != 0:
            raise SystemExit(f"FAILED rc={r.returncode}: {cmd[:120]}")
        return r


log = Log()


def first(pattern):
    hits = sorted(glob.glob(pattern, recursive=True))
    return hits[0] if hits else None
````

### `evalkit/envsetup.py`
ScorerEnv: the scorer's Python environment, offline. Runs in the notebook kernel (Python 3.13), stdlib only.

````python
%%writefile /kaggle/working/evalkit/envsetup.py
"""
ScorerEnv: the scorer's Python environment, offline. Runs in the notebook kernel (Python 3.13), stdlib only.

The wheelhouse is cp312-only and not a consistent dependency set, so the venv is built once on a CPU kernel with
Internet (kaggle/env/build_scorer_env.py -> scorer_env.tgz). Here it is restored, then every wheelhouse wheel whose
version differs from the venv (Oct 9: swegemma 0.2.11, adk-submission 0.2.13; their requirements are unchanged)
is installed with --no-deps, plus the test-only deps of the fastapi tasks (dataset test-deps-wheels).
"""
import json
import os
import pathlib
import re
import sys
import time

from .config import INPUT, SANDBOX_ENV, SANDBOX_PIP, SCORER_SITE, VENV
from .util import first, log

TEST_DEPS = re.compile(r"(inline_snapshot|dirty_equals|asttokens|executing)-")


def canon(name):
    return re.sub(r"[-_.]+", "-", name).lower()


class ScorerEnv:
    def __init__(self, venv=VENV, n_gpus=4):
        self.venv, self.n_gpus = pathlib.Path(venv), n_gpus
        self.python = self.venv / "bin" / "python"

    def check_gpus(self):
        log.sh("nvidia-smi --query-gpu=name,memory.total --format=csv,noheader; nproc; free -g | head -2")
        gpus = [g for g in log.sh("nvidia-smi --query-gpu=name --format=csv,noheader").stdout.split("\n") if g.strip()]
        if len(gpus) < self.n_gpus:  # don't burn quota on the wrong machine
            raise SystemExit(f"need GPU L4 x{self.n_gpus}, got {gpus}")

    def restore(self):
        tgz = first(f"{INPUT}/**/scorer_env.tgz")
        assert tgz, "attach the output of kernel gemma4-scorer-env (kernel_sources)"
        log.sh(f"tar -xzf {tgz} -C /", timeout=1800, check=True)

    def installed(self):
        out = log.sh(f"{self.python} -c 'import importlib.metadata as m, json; "
                     "print(json.dumps({d.metadata[\"Name\"]: d.version for d in m.distributions()}))'",
                     check=True).stdout.strip().splitlines()[-1]
        return {canon(k): v.split("+")[0] for k, v in json.loads(out).items()}

    def stale_wheels(self):
        swegemma = first(f"{INPUT}/**/swegemma-*.whl")
        assert swegemma, "attach dataset metric/gemma-4-developer-agent-wheelhouse"
        have, stale = self.installed(), []
        for w in sorted(pathlib.Path(swegemma).parent.glob("*.whl")):
            if "cutlass" in w.name.lower():
                continue
            n, v = w.name.split("-")[:2]
            v = re.sub(r"\+?cu\d+$", "", v)
            if have.get(canon(n)) != v:
                log(f"wheelhouse {n} {v} != venv {have.get(canon(n))}")
                stale.append(w)
        return stale

    def test_dep_wheels(self):
        found = [pathlib.Path(p) for p in sorted(map(str, INPUT.rglob("*.whl"))) if TEST_DEPS.match(pathlib.Path(p).name)]
        if not found:
            log("WARNING: dataset test-deps-wheels not attached -> ~32 fastapi tasks will be unverifiable")
        return found

    def install(self, wheels):
        stage = pathlib.Path("/tmp/wh_stage")
        stage.mkdir(exist_ok=True)
        for w in wheels:  # pip rejects 'cu128' without '+' in a wheel's version
            name = w.name.replace("cu128", "+cu128") if ("cu128" in w.name and "+" not in w.name) else w.name
            if not (stage / name).exists():
                os.symlink(w, stage / name)
        whls = sorted(str(p) for p in stage.glob("*.whl"))
        if not whls:
            return
        r = log.sh(f"{sys.executable} -m pip --python {self.python} install --no-deps --no-index -q " + " ".join(whls),
                   timeout=1800)
        if r.returncode != 0:  # fallback: wheels are zip archives, unpack into site-packages
            site = log.sh(f"{self.python} -c 'import site; print(site.getsitepackages()[0])'").stdout.strip()
            for w in whls:
                log.sh(f"cd {site} && unzip -oq {w}", check=True)

    def ensure_pip(self):
        """uv venvs have no pip; the sandbox's per-task venvs need `python -m pip` (found via _host_env.pth) to
        install task packages. pip is pure Python, so copy the notebook Python's own pip into the venv (offline)."""
        if log.sh(f"{self.python} -m pip --version").returncode == 0:
            return
        import pip
        src = pathlib.Path(pip.__file__).parent
        site = pathlib.Path(log.sh(f"{self.python} -c 'import site; print(site.getsitepackages()[0])'",
                                   check=True).stdout.strip())
        log.sh(f"cp -r {src} {site}/ && cp -r {src.parent}/pip-*.dist-info {site}/", check=True)
        log.sh(f"{self.python} -m pip --version", check=True)

    def task_sandbox(self):
        """Task sandboxes run on this notebook's Python 3.13 (the scorer's sandbox is python:3.13-slim) with the
        scorer site built by gemma4-devset-build; pip is copied into its own dir (no ensurepip offline)."""
        import pip
        tgz = first(f"{INPUT}/**/scorer_site.tgz")
        assert tgz, "attach the output of kernel gemma4-devset-build (scorer_site.tgz)"
        SCORER_SITE.mkdir(parents=True, exist_ok=True)
        log.sh(f"tar -xzf {tgz} -C {SCORER_SITE}", timeout=1800, check=True)
        SANDBOX_PIP.mkdir(parents=True, exist_ok=True)
        src = pathlib.Path(pip.__file__).parent
        log.sh(f"cp -r {src} {SANDBOX_PIP}/ && cp -r {src.parent}/pip-*.dist-info {SANDBOX_PIP}/", check=True)
        py = os.path.realpath(sys.executable)
        env = {"python": py, "site": str(SCORER_SITE), "pip": str(SANDBOX_PIP)}
        pins = first(f"{INPUT}/**/overlays/pins.json")  # optional: kernel output gemma4-test-overlays
        if pins:
            root = pathlib.Path(pins).parent
            env.update(extras=str(root / "extras"), pins_root=str(root / "pins"), pins=json.loads(pathlib.Path(pins).read_text()))
            log(f"overlays: extras + pins for {len(env['pins'])} tasks from {root}")
        SANDBOX_ENV.write_text(json.dumps(env))
        sp = f"/tmp/sbx_check/lib/python{sys.version_info.major}.{sys.version_info.minor}/site-packages"
        log.sh(f"{py} -m venv --without-pip --clear /tmp/sbx_check && printf '%s\\n%s\\n' {SCORER_SITE} {SANDBOX_PIP} "
               f"> {sp}/_s.pth && /tmp/sbx_check/bin/python3 -c 'import sys, pytest, pip; print(\"task sandbox\", "
               "sys.version.split()[0], \"pytest\", pytest.__version__, \"pip\", pip.__version__)'", check=True)

    def verify(self):
        log.sh(f"{self.python} -c 'import sys, torch, vllm, swegemma, adk_submission, google.adk, "
               "importlib.metadata as m; print(sys.version.split()[0], \"torch\", torch.__version__, \"vllm\", "
               "vllm.__version__, \"swegemma\", m.version(\"swegemma\"), \"adk-submission\", "
               "adk_submission.__version__, \"adk\", google.adk.__version__)'", check=True)
        # import needs pytest's pluggy, which the sandbox venv installs per task; here only check presence
        log.sh(f"{self.python} -c 'import importlib.util as u; "
               "print(\"test deps\", {m: bool(u.find_spec(m)) for m in (\"inline_snapshot\", \"dirty_equals\")})'")

    def prepare_training(self, gpus=True):
        """Scorer venv + the training wheels (peft 0.21.2, bitsandbytes 0.50.2: kernel output gemma4-train-wheels)."""
        t = time.time()
        if gpus:
            self.check_gpus()
        self.restore()
        train = [pathlib.Path(p) for p in sorted(map(str, INPUT.rglob("train_wheels/*.whl")))]
        assert train, "attach the output of kernel gemma4-train-wheels (kernel_sources)"
        self.install(self.stale_wheels() + train)
        self.ensure_pip()
        log.sh(f"{self.python} -c 'import torch, transformers, peft, bitsandbytes; print(\"torch\", torch.__version__, "
               "\"transformers\", transformers.__version__, \"peft\", peft.__version__, \"bnb\", bitsandbytes.__version__)'",
               check=True)
        log(f"training venv ready in {(time.time() - t) / 60:.1f} min")

    def prepare(self, gpus=True):
        t = time.time()
        if gpus:
            self.check_gpus()
        self.restore()
        self.install(self.stale_wheels() + self.test_dep_wheels())
        self.ensure_pip()
        self.task_sandbox()
        self.verify()
        log(f"scorer venv ready in {(time.time() - t) / 60:.1f} min")
````

### `evalkit/inputs.py`
Inputs: where the tasks, snapshots, graphs and served model live under /kaggle/input.

````python
%%writefile /kaggle/working/evalkit/inputs.py
"""Inputs: where the tasks, snapshots, graphs and served model live under /kaggle/input."""
import json
import pathlib
from dataclasses import dataclass

from .config import INPUT
from .util import log


@dataclass
class Inputs:
    tasks_path: pathlib.Path
    data: pathlib.Path        # competition data dir (graphs/, embeddings/, wheels/)
    snapshots: pathlib.Path
    model_path: str

    @classmethod
    def locate(cls, devset=False, need_model=True):
        all_tasks = sorted(INPUT.rglob("tasks.jsonl"))
        comp = [p for p in all_tasks if (p.parent / "graphs").is_dir()]
        assert comp or all_tasks, "competition data not attached"
        data = (comp or all_tasks)[0].parent
        tasks_path = data / "tasks.jsonl"
        if devset:
            dev = [p for p in all_tasks if (p.parent / "devset_report.json").exists()]
            assert dev, "devset=True but the gemma4-devset-build output is not attached"
            tasks_path = dev[0]
        cfgs = sorted(p for p in INPUT.rglob("config.json") if "w4a16" in str(p).lower())
        assert cfgs or not need_model, "attach model google/gemma-4/Other/gemma-4-31b-it-qat-w4a16-ct"
        inp = cls(tasks_path, data, tasks_path.parent / "snapshots", str(cfgs[0].parent) if cfgs else "")
        log("inputs:", inp)
        return inp

    def load_tasks(self):
        return [json.loads(l) for l in open(self.tasks_path) if l.strip()]
````

### `evalkit/submission.py`
Submission: the agent under test, unpacked from submission_src.zip (written by the notebook's inputs cell).

````python
%%writefile /kaggle/working/evalkit/submission.py
"""Submission: the agent under test, unpacked from submission_src.zip (written by the notebook's inputs cell)."""
import shutil
import zipfile

import yaml

from .config import SUBMISSION_ZIP, WORK
from .util import log


class Submission:
    def __init__(self, src_zip=SUBMISSION_ZIP, dest=WORK / "submission_under_test"):
        self.src_zip, self.dir = src_zip, dest

    def unpack(self):
        shutil.rmtree(self.dir, ignore_errors=True)
        with zipfile.ZipFile(self.src_zip) as z:
            z.extractall(self.dir)
        self.fetch_adapters()
        shutil.make_archive(str(WORK / "submission"), "zip", self.dir)  # exactly what would be uploaded
        log("submission files:", sorted(str(p.relative_to(self.dir)) for p in self.dir.rglob("*") if p.is_file()))
        return self

    def fetch_adapters(self):
        """Adapters named in any agent YAML (`adapter: <name>`) but not shipped in the zip are copied from an attached
        training run's output (/kaggle/input/**/adapters/<name>/), so a trained LoRA is evaluated without re-uploading."""
        import re
        from .config import INPUT
        names = {m for p in self.dir.rglob("*.yaml") for m in re.findall(r"^\s*adapter:\s*(\S+)", p.read_text(), re.M)}
        for n in sorted(names):
            dest = self.dir / "adapters" / n
            if (dest / "adapter_model.safetensors").exists():
                continue
            found = sorted(INPUT.glob(f"**/adapters/{n}/adapter_model.safetensors"))
            assert found, f"agent declares adapter {n!r}: not in the submission and no attached adapters/{n}/"
            shutil.copytree(found[0].parent, dest)
            log(f"adapter {n} <- {found[0].parent}")

    def declared_model(self):
        """Exactly one model across all agents (scorer rule); swegemma.models.discovery is gone in 0.2.11."""
        from adk_submission import discover_declared_models
        declared = sorted({m.strip() for m in discover_declared_models(self.dir)})
        assert len(declared) == 1, f"submission must declare exactly one model, got {declared}"
        return declared[0]

    def adapters(self):
        from adk_submission import discover_adapters
        from swegemma.config import ALLOWED_ADAPTER_EXTENSIONS
        return discover_adapters(str(self.dir), adapter_extensions=ALLOWED_ADAPTER_EXTENSIONS)

    def eval_budget(self):
        """eval_config.yaml -> {timeout_seconds, max_time_minutes, max_tool_calls, max_turns}."""
        p = self.dir / "eval_config.yaml"
        ev = (yaml.safe_load(p.read_text()) or {}) if p.exists() else {}
        ev = ev.get("evaluation", ev)
        turns = ev.get("max_turns", ev.get("max_llm_calls"))
        return {"timeout_seconds": ev.get("timeout_seconds"), "max_time_minutes": ev.get("max_time_minutes"),
                "max_tool_calls": ev.get("max_tool_calls"), "max_turns": int(turns) if turns is not None else None}
````

### `evalkit/compat.py`
Local-only shims for running the scorer's subprocess sandbox outside its Docker image. Nothing here changes what the

````python
%%writefile /kaggle/working/evalkit/compat.py
"""
Local-only shims for running the scorer's subprocess sandbox outside its Docker image. Nothing here changes what the
agent sees; each shim exists because a Kaggle notebook is not the scorer's container.

patch_sandbox_venv: the scorer venv's base Python is Debian's python3.12 without ensurepip (python3.12-venv is not
  installed and L4 sessions are offline). Debian's venv then calls sys.exit(1) inside venv.create(with_pip=True), and
  swegemma's SubprocessManager.start only catches Exception, so the SystemExit kills the whole evaluation. Turning it
  into an exception lets swegemma take its own fallback (venv without pip; pip comes from the host site-packages via
  _host_env.pth, see envsetup.ScorerEnv.ensure_pip).
"""
import contextvars
import venv

from .util import log

TASK = contextvars.ContextVar("evalkit_task", default=None)  # instance_id of the task whose sandbox is being started


def patch_sandbox_venv():
    if getattr(venv.create, "_evalkit", False):
        return
    orig = venv.create

    def create(*a, **kw):
        try:
            return orig(*a, **kw)
        except SystemExit as e:  # Debian venv without ensurepip
            raise RuntimeError(f"venv.create exited ({e}); ensurepip unavailable") from None

    create._evalkit = True
    venv.create = create
    log("compat: venv.create SystemExit -> RuntimeError (swegemma falls back to a venv without pip)")


def use_scorer_sandbox(data_dir):
    """Make each task sandbox look like the scorer's Docker sandbox (python:3.13-slim + Dockerfile pip packages +
    every competition wheel unpacked + sandbox/setup.py --fast-path), as in kaggle/devset/build_devset.py, where gold
    patches then pass 54/54 on the exec-graph tasks.

    - In subprocess mode swegemma skips test-dependency setup and task venvs inherit the HOST site-packages (here the
      harness's Py3.12 venv: no pytest, wrong versions). Instead: no system site-packages, a fresh Python 3.13 venv
      (the image's interpreter, matching python:3.13-slim and the cp313 wheels in scorer_site), and a .pth pointing at
      the restored scorer site (gemma4-devset-build output) plus a pip-only dir (Debian Python has no ensurepip).
    - install_test_dependencies runs the competition's sandbox/setup.py, like the scorer's container setup.
    """
    import json
    import pathlib
    import subprocess

    import swegemma.evaluate as ev
    import swegemma.harness.container_setup as cs
    import swegemma.harness.verification as vf
    import swegemma.sandbox.subprocess as sp

    from .config import SANDBOX_ENV

    env = json.loads(SANDBOX_ENV.read_text())
    if not getattr(sp.SubprocessManager, "_evalkit", False):
        orig_init, orig_start = sp.SubprocessManager.__init__, sp.SubprocessManager.start

        def init(self, *a, **kw):
            kw["system_site_packages"] = False
            orig_init(self, *a, **kw)

        def start(self, *a, **kw):
            sid = orig_start(self, *a, **kw)
            vdir = pathlib.Path(self._sandboxes[sid]["venv"])
            subprocess.run([env["python"], "-m", "venv", "--without-pip", "--clear", str(vdir)], check=True,
                           capture_output=True)
            # search order: this task's version pins -> test-only extras -> scorer site -> pip
            path = [f"{env['pins_root']}/{d}" for d in env.get("pins", {}).get(TASK.get(), [])]
            path += [env["extras"]] if env.get("extras") else []
            path += [env["site"], env["pip"]]
            for spd in vdir.glob("lib/python*/site-packages"):
                (spd / "_scorer_site.pth").write_text("\n".join(path) + "\n")
            return sid

        sp.SubprocessManager.__init__, sp.SubprocessManager.start = init, start
        sp.SubprocessManager._evalkit = True

    if not getattr(ev.Evaluator, "_evalkit_task", False):  # asyncio.to_thread copies the context into start()
        orig_eval = ev.Evaluator.evaluate_task

        async def evaluate_task(self, task, *a, **kw):
            TASK.set(task.instance_id)
            return await orig_eval(self, task, *a, **kw)

        ev.Evaluator.evaluate_task, ev.Evaluator._evalkit_task = evaluate_task, True

    setup_py = pathlib.Path(data_dir) / "sandbox" / "setup.py"
    assert setup_py.exists(), f"missing {setup_py}"
    orig_deps = cs.install_test_dependencies

    def install_test_dependencies(docker, container_id, repo="", *, fast_path=True, config=None):
        if type(docker).__name__ not in {"SubprocessManager", "SubprocessSandbox"}:
            return orig_deps(docker, container_id, repo, fast_path=fast_path, config=config)
        flag = "--fast-path" if fast_path else "--no-fast-path"
        docker.exec(container_id, f"python3 {setup_py} {flag} --workspace /workspace {repo}".strip(), timeout=600)

    for mod in (cs, vf, ev):
        if hasattr(mod, "install_test_dependencies"):
            mod.install_test_dependencies = install_test_dependencies
    log(f"compat: task sandboxes = {env['python']} venv + scorer site {env['site']} + {setup_py.name}"
        + (f" + extras + pins for {len(env.get('pins', {}))} tasks" if env.get("extras") else ""))
````

### `evalkit/server.py`
ScorerServer: vLLM 0.19.1 with the scorer's settings, and the agent sandbox made offline.

````python
%%writefile /kaggle/working/evalkit/server.py
"""ScorerServer: vLLM 0.19.1 with the scorer's settings, and the agent sandbox made offline."""
import os
import time

from .config import TARGET_MODEL
from .util import log

SCORER_ENV = {"LITELLM_LOCAL_MODEL_COST_MAP": "True", "TRANSFORMERS_NO_TF": "1", "VLLM_WORKER_MULTIPROC_METHOD": "spawn",
              "VLLM_MEMORY_PROFILER_ESTIMATE_CUDAGRAPHS": "1", "VLLM_ENGINE_READY_TIMEOUT_S": "1200",
              "VLLM_NO_USAGE_STATS": "1", "OTEL_SDK_DISABLED": "true",
              "PYTORCH_CUDA_ALLOC_CONF": "expandable_segments:True"}


def set_scorer_env():
    os.environ.update(SCORER_ENV)


def make_sandbox_offline():
    """Dead proxy for everything except localhost, like the scorer's network_mode=none."""
    for k in ("http_proxy", "https_proxy", "HTTP_PROXY", "HTTPS_PROXY", "ALL_PROXY", "all_proxy"):
        os.environ[k] = "http://127.0.0.1:9"
    os.environ["NO_PROXY"] = os.environ["no_proxy"] = "localhost,127.0.0.1"
    os.environ["PIP_NO_INDEX"] = "1"


class ScorerServer:
    def __init__(self, model_path, adapters):
        import torch
        from adk_submission import VllmConfig, VllmServer
        self.server = VllmServer(VllmConfig(
            model=model_path, port=8000, host="127.0.0.1", tool_call_parser="gemma4", reasoning_parser="gemma4",
            default_chat_template_kwargs={"enable_thinking": True}, max_model_len=32768,
            dtype="bfloat16" if torch.cuda.is_bf16_supported() else "auto", gpu_memory_utilization=0.80,
            enable_auto_tool_choice=True, enable_lora=True, max_loras=8, max_lora_rank=128,
            tensor_parallel_size=4, startup_timeout=60 * 25), adapter_manifest=adapters)

    def start(self):
        t = time.time()
        self.server.start()
        log(f"vLLM up on {self.server.base_url} after {time.time() - t:.0f}s")
        return self

    def registry(self, declared_model):
        """LiteLLM clients for the declared model name (and the target name) plus one per adapter."""
        return self.server.create_model_registry(aliases=[declared_model, TARGET_MODEL], model_prefix="openai/",
                                                 api_key="EMPTY")

    def stop(self):
        try:
            self.server.stop()
        except Exception:  # noqa: BLE001
            pass
````

### `evalkit/runner.py`
Task order + ChunkedEvaluation: the official swegemma Evaluator, one task at a time, in budgeted chunks.

````python
%%writefile /kaggle/working/evalkit/runner.py
"""Task order + ChunkedEvaluation: the official swegemma Evaluator, one task at a time, in budgeted chunks."""
import asyncio
import json
import random
import time

from .config import VALID_TASKS, WORK
from .util import log


def stratified_order(tasks, seed=0):
    """Shuffle within each repo, then interleave repos (largest first) so any prefix is a stratified sample."""
    rng = random.Random(seed)
    by_repo = {}
    for tk in tasks:
        by_repo.setdefault(tk["repo"], []).append(tk["instance_id"])
    pools = []
    for v in by_repo.values():
        rng.shuffle(v)
        pools.append(list(v))
    ids, total = [], sum(len(p) for p in pools)
    while len(ids) < total:
        for p in sorted(pools, key=len, reverse=True):
            if p:
                ids.append(p.pop())
    return ids


class ChunkedEvaluation:
    def __init__(self, settings, inputs, submission, models, adapters):
        self.s, self.inp, self.sub = settings, inputs, submission
        self.models, self.adapters = models, adapters
        self.budget = submission.eval_budget()
        self.root = WORK / settings.label
        ids = list(settings.task_ids) if settings.task_ids else stratified_order(inputs.load_tasks(), settings.seed)
        if settings.valid_only:  # tasks the scorer env can verify (gold passes, empty fails): see preflight sweep
            valid = set(json.loads(VALID_TASKS.read_text()))
            ids = [i for i in ids if i in valid]
        self.ids = ids[:settings.max_tasks] if settings.max_tasks else ids
        (WORK / "task_ids_order.json").write_text(json.dumps(self.ids, indent=1))
        log(f"{len(self.ids)} tasks x {settings.repeats} rep(s), budget {self.budget}")

    def config(self, chunk, results_dir):
        """EvalConfig with the scorer's limits, context cache and compaction (interval 15 / threshold 32768)."""
        from google.adk.agents.context_cache_config import ContextCacheConfig
        from google.adk.apps._configs import EventsCompactionConfig
        from swegemma.config import EvalConfig, build_submission_limits
        limits, gen = build_submission_limits()
        return EvalConfig(
            tasks_path=self.inp.tasks_path, snapshots_dir=self.inp.snapshots, results_dir=results_dir,
            submission_dir=self.sub.dir, models=self.models, sandbox="subprocess", wheels_dir=self.inp.data / "wheels",
            graph_dir=str(self.inp.data / "graphs"), embeddings_dir=str(self.inp.data / "embeddings"),
            **self.budget, limits=limits, generation_constraints=gen, adapter_manifest=self.adapters,
            context_cache_config=ContextCacheConfig(min_tokens=2048, ttl_seconds=1800, cache_intervals=10),
            events_compaction_config=EventsCompactionConfig(compaction_interval=15, overlap_size=2,
                                                            token_threshold=32768, event_retention_size=5),
            task_ids=chunk, concurrency=self.s.concurrency, display_mode="quiet", verbose=False)

    def task_min(self):
        """The submission's per-task agent time limit (eval_config.yaml max_time_minutes)."""
        return float(self.budget.get("max_time_minutes") or 4.5)

    def run_chunk(self, rep, start):
        from swegemma.evaluate import Evaluator
        chunk = self.ids[start:start + self.s.chunk_size]
        t = time.time()
        try:
            res = asyncio.run(asyncio.wait_for(
                Evaluator(self.config(chunk, self.root / f"rep{rep}" / f"chunk_{start:03d}")).run(),
                timeout=max(self.s.chunk_timeout_min, len(chunk) * (self.task_min() + 3)) * 60))
            log(f"rep {rep} chunk {start // self.s.chunk_size}: {repr(res)[:300]} ({(time.time() - t) / 60:.1f} min)")
        except Exception as e:  # noqa: BLE001
            log(f"rep {rep} chunk {start // self.s.chunk_size} FAILED: {type(e).__name__}: {e}")

    def healthy(self, chunk_dir):
        """After the first chunk: abort on systemic failure instead of burning a 10 h session on garbage."""
        p = chunk_dir / "task_results.jsonl"
        rows = [json.loads(l) for l in open(p) if l.strip()] if p.exists() else []
        traces = list((chunk_dir / "traces").glob("*.json"))
        checks = {"has_results": bool(rows),
                  "not_all_errors": any(not r.get("error") for r in rows),
                  "model_calls_tools": any((r.get("tool_calls") or 0) > 0 for r in rows),
                  "traces_saved": bool(traces),
                  # task_results.jsonl has no test output: it is in test_outputs/<id>.log; exit code -1 = not run
                  "tests_ran": any((r.get("test_exit_code") if r.get("test_exit_code") is not None else -1) >= 0
                                   for r in rows) and any((chunk_dir / "test_outputs").glob("*.log"))}
        log("HEALTH after first chunk:", json.dumps(checks),
            "| errors:", [str(r.get("error"))[:120] for r in rows if r.get("error")][:3])
        return all(checks.values())

    def run(self):
        """Rep-major: rep 0 covers every task before rep 1 starts, so a budget stop still leaves a full eval."""
        for rep in range(self.s.repeats):
            for start in range(0, len(self.ids), self.s.chunk_size):
                n = len(self.ids[start:start + self.s.chunk_size])
                if self.s.hours_left() < n * max(self.s.est_task_min, self.task_min() + 1) / 60 / self.s.concurrency:
                    log(f"STOP: {self.s.hours_left():.2f} h left (rep {rep}, task {start})")
                    return
                self.run_chunk(rep, start)
                if rep == 0 and start == 0 and not self.healthy(self.root / "rep0" / "chunk_000"):
                    log("ABORT: first chunk failed the health check; fix before a full run")
                    return
````

### `evalkit/report.py`
Report: merge chunk outputs per rep, pass@k summary, trace analysis, and one tarball for download.

````python
%%writefile /kaggle/working/evalkit/report.py
"""Report: merge chunk outputs per rep, pass@k summary, trace analysis, and one tarball for download."""
import json
import shutil
import sys

from .config import ANALYZER, WORK
from .util import log

PARTS = ("traces", "patches", "test_outputs", "logs")


class Report:
    def __init__(self, settings, inputs):
        self.s, self.inp = settings, inputs
        self.root = WORK / settings.label

    def merge_rep(self, rep_dir):
        merged = rep_dir / "merged"
        for part in PARTS:
            (merged / part).mkdir(parents=True, exist_ok=True)
        with open(merged / "task_results.jsonl", "w") as out:
            for ch in sorted(rep_dir.glob("chunk_*")):
                if (ch / "task_results.jsonl").exists():
                    out.write((ch / "task_results.jsonl").read_text())
                for part in PARTS:
                    for f in (ch / part).glob("*"):
                        shutil.copy(f, merged / part / f.name)
                shutil.rmtree(ch)
        return merged, [json.loads(l) for l in open(merged / "task_results.jsonl") if l.strip()]

    def analyze(self, merged, out):
        if ANALYZER.exists():
            log.sh(f"{sys.executable} {ANALYZER} --results {merged} --tasks {self.inp.tasks_path} --out {out}")

    def build(self):
        summary = {"name": self.s.name, "devset": self.s.devset, "reps": {}}
        solved = {}
        for rep_dir in sorted(self.root.glob("rep*")):
            merged, rows = self.merge_rep(rep_dir)
            for r in rows:
                solved.setdefault(r["instance_id"], []).append(bool(r.get("resolved")))
            summary["reps"][rep_dir.name] = {"resolved": sum(bool(r.get("resolved")) for r in rows), "n": len(rows)}
            self.analyze(merged, rep_dir / "analysis")
        summary["pass_at_k"] = {"tasks": len(solved), "any": sum(any(v) for v in solved.values()),
                                "all": sum(all(v) for v in solved.values())}
        (WORK / "summary.json").write_text(json.dumps(summary, indent=1))
        log("SUMMARY", json.dumps(summary))
        return summary

    def package(self):
        log.sh(f"cd {WORK} && tar -czf {self.s.label}.tgz {self.s.label} task_ids_order.json run.log summary.json "
               f"&& rm -rf {self.s.label} submission_under_test; ls -la {WORK}")
````

### `evalkit/preflight.py`
Preflight (CPU, offline, no model): the official Evaluator with a fake agent that submits the GOLD patch for tasks

````python
%%writefile /kaggle/working/evalkit/preflight.py
"""
Preflight (CPU, offline, no model): the official Evaluator with a fake agent that submits the GOLD patch for tasks
known to pass in the scorer env (exec-graph run, 54/54), and an EMPTY patch for one of them. Proves task sandboxes,
test deps, verification and result files before a GPU session is spent. Expect gold -> resolved, empty -> not.
Run inside the venv: python -m evalkit.preflight
"""
import asyncio
import json
import pathlib

import pandas as pd

from .compat import patch_sandbox_venv, use_scorer_sandbox
from .config import WORK, Settings
from .inputs import Inputs
from .server import make_sandbox_offline, set_scorer_env
from .util import log

GOLD_OK = ["fastapi_14258", "requests_6644", "rich_2725", "fastapi_14372", "requests_7205", "rich_2943"]
EMPTY = "rich_3006"  # gold-passing task submitted with no patch: must NOT resolve


def gold_patches(inp):
    """Gold patches: secret/perfect_submission.* anywhere under /kaggle/input, else the tasks' own `patch` field.
    Also reports where the verifier's test patches (secret/solution.*) live: without them nothing can be verified."""
    from .config import INPUT
    log("data dir:", sorted(p.name for p in inp.data.iterdir()))
    log("task keys:", sorted(inp.load_tasks()[0]))
    sols = sorted(INPUT.rglob("solution.parquet")) + sorted(INPUT.rglob("solution.csv"))
    log("secret solution files:", [str(p) for p in sols] or "NONE (verifier has no test patches)")
    for f in sorted(INPUT.rglob("perfect_submission.parquet")) + sorted(INPUT.rglob("perfect_submission.csv")):
        df = pd.read_parquet(f) if f.suffix == ".parquet" else pd.read_csv(f)
        log("gold from", f)
        return {str(r.get("id") or r.get("instance_id")): r.get("prediction") or r.get("patch")
                for r in df.to_dict(orient="records")}
    log("gold from tasks.jsonl `patch`")
    return {t["instance_id"]: t.get("patch") or "" for t in inp.load_tasks()}


def verify(inp, patches, label):
    """Official Evaluator with a fake agent that returns patches[instance_id]; -> {id: row}."""
    from adk_eval_core.tracing import SessionTrace
    from adk_submission import ModelRegistry
    from swegemma.config import EvalConfig
    from swegemma.evaluate import Evaluator

    async def fake_agent(self, task, snapshot_path, task_index, total_tasks, **kw):
        return patches[task.instance_id], None, SessionTrace()

    Evaluator._run_agent_sandbox = fake_agent
    dummy = WORK / "preflight_submission"
    dummy.mkdir(exist_ok=True)
    cfg = EvalConfig(tasks_path=inp.tasks_path, snapshots_dir=inp.snapshots, results_dir=WORK / f"preflight_{label}",
                     submission_dir=dummy, models=ModelRegistry(), sandbox="subprocess", wheels_dir=inp.data / "wheels",
                     graph_dir=str(inp.data / "graphs"), embeddings_dir=str(inp.data / "embeddings"),
                     task_ids=sorted(patches), concurrency=4, display_mode="quiet", max_time_minutes=30,
                     timeout_seconds=600)
    res = asyncio.run(Evaluator(cfg).run())
    return {r.instance_id: {"resolved": bool(r.resolved), "exit": r.test_exit_code, "error": r.error,
                            "tail": (r.test_output or "")[-600:]} for r in res.task_results}


def main():
    s = Settings.load()
    set_scorer_env()
    patch_sandbox_venv()
    inp = Inputs.locate(False, need_model=False)
    use_scorer_sandbox(inp.data)
    make_sandbox_offline()
    gold = gold_patches(inp)
    ids = sorted(gold) if s.sweep else GOLD_OK + [EMPTY]
    g = verify(inp, {i: gold[i] for i in ids if i != EMPTY or s.sweep} | ({} if s.sweep else {EMPTY: ""}), "gold")
    e = verify(inp, {i: "" for i in ids}, "empty") if s.sweep else {EMPTY: g.pop(EMPTY)}
    (WORK / "preflight.json").write_text(json.dumps({"gold": g, "empty": e}, indent=1))
    if s.sweep:
        valid = sorted(i for i in ids if g.get(i, {}).get("resolved") and not e.get(i, {}).get("resolved", True))
        (WORK / "valid_tasks.json").write_text(json.dumps(valid, indent=1))
        by_repo = {}
        for i in ids:
            r = by_repo.setdefault(i.rsplit("_", 1)[0], [0, 0, 0])
            r[0] += 1
            r[1] += bool(g.get(i, {}).get("resolved"))
            r[2] += i in valid
        log(f"SWEEP: valid {len(valid)}/{len(ids)} (gold passes and empty fails); per repo [n, gold, valid]: {by_repo}")
        return
    gold_ok = sum(g.get(i, {}).get("resolved", False) for i in GOLD_OK)
    empty_ok = not e.get(EMPTY, {}).get("resolved", True)
    verdict = "PASS" if gold_ok >= len(GOLD_OK) - 1 and empty_ok else "FAIL"
    log(f"PREFLIGHT {verdict}: gold resolved {gold_ok}/{len(GOLD_OK)}, empty patch unresolved={empty_ok}")
    for i, r in (g | e).items():
        log(i, r["resolved"], r["exit"], r["error"], r["tail"][-200:].replace("\n", " | "))


if __name__ == "__main__":
    main()
````

### `evalkit/overlays.py`
Overlays: extra packages for the task sandboxes so more public tasks are verifiable (CPU kernel, Internet ON).

````python
%%writefile /kaggle/working/evalkit/overlays.py
"""
Overlays: extra packages for the task sandboxes so more public tasks are verifiable (CPU kernel, Internet ON).

Why: 75/129 public tasks fail even with the gold patch in the scorer env (Oct 14 sweep): fastapi tests import
inline_snapshot / dirty_equals (31), python-multipart (4), old fastapi commits call Starlette APIs removed in 1.x (24), requests tests need
the pytest-httpbin plugin (7), rich colour snapshots changed with newer Pygments (7), `attr` missing (1). The leaderboard
scores private repos, so these public tasks only need to measure the agent reliably, not match the scorer.
Output (/kaggle/working/overlays):
  extras/      test-only packages for every task, resolved against the scorer site's versions (packages the site
               already has are removed, so only new ones are added)
  pins/<pkg>-<version>/   one package each, --no-deps
  pins.json    {instance_id: [pin dir, ...]}: only tasks that are invalid today, and only for PINNED repos; each
               package pinned to its latest release before the task's created_at (the env the tests were written in)
  report.json  what was built
Used by compat.use_scorer_sandbox (sandbox path: task pins -> extras -> scorer site -> pip); the gold/empty sweep
then decides which tasks are valid.
Run: python -m evalkit.overlays (no venv needed: plain pip in the notebook's Python 3.13, same as the sandbox).
"""
import datetime as dt
import json
import re
import shutil
import subprocess
import sys
import urllib.request

from .config import INPUT, SCORER_SITE, VALID_TASKS, WORK
from .util import first, log

EXTRAS = ["inline-snapshot", "dirty-equals", "pytest-httpbin", "httpbin", "attrs", "python-multipart"]
PINNED = {"fastapi": ["starlette"], "rich": ["pygments"]}
OUT = WORK / "overlays"


def canon(name):
    return re.sub(r"[-_.]+", "-", name).lower()


class OverlayBuilder:
    def __init__(self, tasks, valid):
        self.tasks, self.valid = tasks, set(valid)
        self.report = {"extras": [], "pins": {}, "errors": []}

    def pip(self, *args):
        r = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--disable-pip-version-check", *args],
                           capture_output=True, text=True)
        if r.returncode:
            self.report["errors"].append(" ".join(args)[:200] + " -> " + r.stderr[-400:])
        return r.returncode == 0

    def site_versions(self):
        """{canonical name: version} of the scorer site (restored from gemma4-devset-build's scorer_site.tgz)."""
        tgz = first(f"{INPUT}/**/scorer_site.tgz")
        assert tgz, "attach the output of kernel gemma4-devset-build (scorer_site.tgz)"
        SCORER_SITE.mkdir(parents=True, exist_ok=True)
        log.sh(f"tar -xzf {tgz} -C {SCORER_SITE}", timeout=1800, check=True)
        have = {}
        for d in SCORER_SITE.glob("*.dist-info"):
            m = re.match(r"(.+?)-([^-]+)\.dist-info$", d.name)
            if m:
                have[canon(m.group(1))] = m.group(2)
        return have

    def build_extras(self):
        have = self.site_versions()
        cons = WORK / "site_constraints.txt"
        cons.write_text("".join(f"{n}=={v}\n" for n, v in sorted(have.items())))
        ext = OUT / "extras"
        shutil.rmtree(ext, ignore_errors=True)
        for pkg in EXTRAS:  # one at a time: a resolution conflict in one must not block the others
            how = "constrained" if self.pip("--target", str(ext), "-c", str(cons), pkg) else (
                "unconstrained" if self.pip("--target", str(ext), pkg) else "FAILED")
            self.report.setdefault("extras_install", {})[pkg] = how
            log(f"extra {pkg}: {how}")
        for d in list(ext.glob("*.dist-info")):  # drop what the site already provides (same pinned version)
            name = canon(re.match(r"(.+?)-[^-]+\.dist-info$", d.name).group(1))
            if name in have:
                for line in (d / "RECORD").read_text().splitlines() if (d / "RECORD").exists() else []:
                    top = line.split(",")[0].split("/")[0]
                    p = (ext / top).resolve()
                    # only real top-level entries inside extras/ (RECORD also lists scripts as ../../bin/...)
                    if top in ("", ".", "..") or top.endswith(".dist-info") or p.parent != ext.resolve():
                        continue
                    shutil.rmtree(p, ignore_errors=True) if p.is_dir() else p.unlink(missing_ok=True)
                shutil.rmtree(d, ignore_errors=True)
        self.report["extras"] = sorted(d.name for d in ext.glob("*.dist-info"))
        log("extras:", self.report["extras"])

    @staticmethod
    def releases(pkg):
        """[(version, upload time)] of non-yanked final releases, from PyPI's JSON API."""
        data = json.load(urllib.request.urlopen(f"https://pypi.org/pypi/{pkg}/json", timeout=60))
        out = []
        for v, files in data["releases"].items():
            if not files or re.search(r"[a-zA-Z]", v) or all(f.get("yanked") for f in files):
                continue
            t = min(dt.datetime.fromisoformat(f["upload_time_iso_8601"].replace("Z", "+00:00")) for f in files)
            out.append((v, t))
        return sorted(out, key=lambda x: x[1])

    def build_pins(self):
        rel = {p: self.releases(p) for ps in PINNED.values() for p in ps}
        pins = {}
        for t in self.tasks:
            iid, repo_short = t["instance_id"], t["instance_id"].rsplit("_", 1)[0]
            if iid in self.valid or repo_short not in PINNED:
                continue
            when = dt.datetime.fromisoformat(str(t.get("created_at")).replace("Z", "+00:00"))
            if when.tzinfo is None:
                when = when.replace(tzinfo=dt.timezone.utc)
            for pkg in PINNED[repo_short]:
                before = [v for v, up in rel[pkg] if up <= when]
                if not before:
                    continue
                name = f"{pkg}-{before[-1]}"
                d = OUT / "pins" / name
                if not d.exists() and not self.pip("--no-deps", "--target", str(d), f"{pkg}=={before[-1]}"):
                    continue
                pins.setdefault(iid, []).append(name)
        (OUT / "pins.json").write_text(json.dumps(pins, indent=1))
        self.report["pins"] = {i: v for i, v in sorted(pins.items())}
        log(f"pins for {len(pins)} tasks:", sorted({n for v in pins.values() for n in v}))

    def build(self):
        OUT.mkdir(parents=True, exist_ok=True)
        self.build_extras()
        self.build_pins()
        (OUT / "report.json").write_text(json.dumps(self.report, indent=1))
        log("overlays built:", json.dumps({k: (len(v) if isinstance(v, (list, dict)) else v)
                                          for k, v in self.report.items()}))


def main():
    tasks_path = next(p for p in sorted(INPUT.rglob("tasks.jsonl")) if (p.parent / "graphs").is_dir())
    tasks = [json.loads(l) for l in open(tasks_path) if l.strip()]
    valid = json.loads(VALID_TASKS.read_text()) if VALID_TASKS.exists() else []
    log(f"{len(tasks)} tasks, {len(valid)} valid today")
    OverlayBuilder(tasks, valid).build()


if __name__ == "__main__":
    main()
````

### `evalkit/gate.py`
Behaviour gate (docs/finetune_strategy_plan.md §4): before any task evaluation, ask base v7 and each candidate

````python
%%writefile /kaggle/working/evalkit/gate.py
"""
Behaviour gate (docs/finetune_strategy_plan.md §4): before any task evaluation, ask base v7 and each candidate
adapter for their NEXT call at fixed decision points cut from v7's own runs (posttrain/gate_prefixes.py), with v7's
exact system prompt and tool declarations, and score the kind of call each one makes.

  file_in_view -> edit            test_failed -> edit or read (not submit, not the same test again)
  test_passed  -> submit_patch    early       -> read/search (not pytest, not submit)
  every response: exactly one tool call
One vLLM server (the scorer's settings) serves the base model and every adapter at once; n samples per prefix at
v7's sampling settings. A candidate passes when its edit rate at file_in_view and its overall pass rate are >= base,
and it never batches calls. Output: /kaggle/working/gate_report.json.
Run inside the venv: python -m evalkit.gate
"""
import json
import pathlib
import re
import shutil
import urllib.request

from .config import INPUT, WORK, Settings
from .inputs import Inputs
from .server import ScorerServer, set_scorer_env
from .util import log

PREFIXES = WORK / "gate_prefixes.json"
GATE_SYSTEM = WORK / "gate_system.md"
GATE_TOOLS = WORK / "gate_tools.json"


def kind(call):
    f = call.get("function", {})
    if f.get("name") != "run_command":
        return f.get("name") or "none"
    try:
        args = json.loads(f.get("arguments") or "{}")
    except Exception:  # noqa: BLE001
        return "malformed"
    c = re.sub(r"^(#[^\n]*\n)+", "", str(args.get("command", "")).strip())  # a leading reasoning comment is fine
    c = re.sub(r"^cd /workspace\s*&&\s*", "", c)
    if "PYEOF" in c and "write_text(" in c:
        return "edit"
    if re.match(r"(python3? -m )?pytest|python3? -m unittest", c):
        return "test"
    if re.match(r"(sed -n|cat|grep|find|ls|rg|head|tail)\b", c):  # "cat \b" never matched `cat -n` (fixed Oct 18)
        return "read"
    return "other"


def passes(typ, k, prev_cmd, cmd):
    if typ == "file_in_view":
        return k == "edit"
    if typ == "test_failed":
        return k in ("edit", "read") and cmd != prev_cmd
    if typ == "test_passed":
        return k == "submit_patch"
    return k == "read"  # early


def as_api(messages):
    """OpenAI chat format: earlier tool-call arguments must be JSON strings (the prefixes store dicts)."""
    out = []
    for m in messages:
        m = dict(m)
        if m.get("tool_calls"):
            m["tool_calls"] = [{**tc, "function": {**tc["function"], "arguments": tc["function"]["arguments"]
                                if isinstance(tc["function"]["arguments"], str)
                                else json.dumps(tc["function"]["arguments"], ensure_ascii=False)}}
                               for tc in m["tool_calls"]]
            m.setdefault("content", None)
        out.append(m)
    return out


def chat(base_url, model, messages, tools, n):
    body = {"model": model, "messages": messages, "tools": tools, "tool_choice": "auto", "n": n,
            "temperature": 0.2, "top_p": 0.95, "top_k": 40, "max_tokens": 1024,
            "chat_template_kwargs": {"enable_thinking": False}}
    req = urllib.request.Request(f"{base_url}/chat/completions", data=json.dumps(body).encode(),
                                 headers={"Content-Type": "application/json"})
    return json.load(urllib.request.urlopen(req, timeout=600))


def main():
    s = Settings.load()
    set_scorer_env()
    inp = Inputs.locate(False)
    names = [n for n in (s.gate_adapters or []) if n]
    sub = WORK / "gate_submission"
    shutil.rmtree(sub, ignore_errors=True)
    for n in names:  # adapters from attached training outputs: /kaggle/input/**/adapters/<name>/
        found = sorted(INPUT.glob(f"**/adapters/{n}/adapter_model.safetensors"))
        assert len(found) == 1, f"adapter {n!r}: expected exactly one attached copy, found {found}"
        shutil.copytree(found[0].parent, sub / "adapters" / n)
    from adk_submission import discover_adapters
    from swegemma.config import ALLOWED_ADAPTER_EXTENSIONS
    adapters = discover_adapters(str(sub), adapter_extensions=ALLOWED_ADAPTER_EXTENSIONS) if names else None
    server = ScorerServer(inp.model_path, adapters).start()
    base_url = server.server.base_url
    served = [m["id"] for m in json.load(urllib.request.urlopen(f"{base_url}/models"))["data"]]
    base_id = next(m for m in served if m not in names)
    log("served:", served, "| base:", base_id)
    system = GATE_SYSTEM.read_text()
    tools = json.loads(GATE_TOOLS.read_text())
    prefixes = json.loads(PREFIXES.read_text())
    report = {"models": {}, "n": s.gate_samples, "prefixes": len(prefixes)}
    for label, model in [("base", base_id)] + [(n, n) for n in names]:
        rows, by_type = [], {}
        for p in prefixes:
            msgs = as_api([{"role": "system", "content": system}] + p["messages"])
            prev = next((m["tool_calls"][0]["function"]["arguments"].get("command") for m in reversed(p["messages"])
                         if m.get("tool_calls")), None)
            try:
                out = chat(base_url, model, msgs, tools, s.gate_samples)
            except Exception as e:  # noqa: BLE001
                rows.append({"id": p["id"], "error": f"{type(e).__name__}: {e}"[:200]})
                continue
            for ch in out["choices"]:
                calls = ch["message"].get("tool_calls") or []
                k = kind(calls[0]) if calls else "no_call"
                cmd = None
                if calls:
                    try:
                        cmd = json.loads(calls[0]["function"].get("arguments") or "{}").get("command")
                    except Exception:  # noqa: BLE001
                        cmd = None
                ok = len(calls) == 1 and passes(p["type"], k, prev, cmd)
                rows.append({"id": p["id"], "type": p["type"], "kind": k, "n_calls": len(calls), "pass": ok,
                             "call": (calls[0]["function"].get("name", "") + " " + (cmd or ""))[:300] if calls
                             else (ch["message"].get("content") or "")[:300]})
                t = by_type.setdefault(p["type"], {"pass": 0, "n": 0, "kinds": {}})
                t["pass"] += ok
                t["n"] += 1
                t["kinds"][k] = t["kinds"].get(k, 0) + 1
        scored = [r for r in rows if "pass" in r]
        report["models"][label] = {
            "pass_rate": round(sum(r["pass"] for r in scored) / max(1, len(scored)), 3),
            "edit_rate_file_in_view": round(by_type.get("file_in_view", {}).get("pass", 0)
                                            / max(1, by_type.get("file_in_view", {}).get("n", 0)), 3),
            "batched_responses": sum(r["n_calls"] > 1 for r in scored),
            "no_call_responses": sum(r["n_calls"] == 0 for r in scored),
            "errors": sum("error" in r for r in rows), "by_type": by_type, "rows": rows}
        log(f"GATE {label}: pass {report['models'][label]['pass_rate']} | edit@file_in_view "
            f"{report['models'][label]['edit_rate_file_in_view']} | batched {report['models'][label]['batched_responses']}")
    base = report["models"]["base"]
    for n in names:
        m = report["models"][n]
        m["verdict"] = "PASS" if (m["edit_rate_file_in_view"] >= base["edit_rate_file_in_view"]
                                  and m["pass_rate"] >= base["pass_rate"] and m["batched_responses"] == 0
                                  and m["errors"] == 0) else "FAIL"
        log(f"VERDICT {n}: {m['verdict']}")
    (WORK / "gate_report.json").write_text(json.dumps(report, indent=1))
    server.stop()


if __name__ == "__main__":
    main()
````

````python
%%writefile /kaggle/working/evalkit/necessity.py
"""
A3/A4 (CPU, offline, no model): test-necessity of gold function groups and the completion counterfactual, through
the official Evaluator with a fake agent, exactly like the preflight sweep. Pre-registered in docs/experiments.md.
  A3  new-60 tasks whose gold touches >= 2 function groups: gold, empty, and gold minus group i (loo<i>).
  A4  every unresolved, non-empty agent patch (embedded runs) that touches >= 1 gold group and misses >= 1:
      agent   = the agent's patch alone (control, should stay unresolved)
      comp    = agent patch + gold hunks of the missed groups + gold module-level hunks (the completion)
      missed  = those gold hunks alone (control: was the agent's own part needed?)
Function groups use the exec-graph rule: each gold hunk maps to the innermost function(s) enclosing its changed lines.
Inputs: /kaggle/working/a3a4_payload.tgz (agent patches + per_task rows + old-54 LOO results), embedded by
make_eval_notebook.py --necessity. Output: /kaggle/working/necessity/*.jsonl|json and necessity_report.json.
Run inside the venv: python -m evalkit.necessity
"""
import ast
import json
import pathlib
import re
import shutil
import subprocess
import tarfile
import tempfile
import time

from .compat import patch_sandbox_venv, use_scorer_sandbox
from .config import VALID_TASKS, WORK, Settings
from .inputs import Inputs
from .preflight import gold_patches, verify
from .server import make_sandbox_offline, set_scorer_env
from .util import log

OUT = WORK / "necessity"
WS = pathlib.Path("/tmp/a3a4_ws")
PAYLOAD = WORK / "a3a4_payload.tgz"
IN = WORK / "a3a4_in"


# ------------------------------------------------------------------ gold function groups (exec-graph rule)
def split_patch(patch):
    """-> [(file_header_lines, [(old_start, old_len, hunk_lines)])] per file"""
    files, cur, hunk = [], None, None
    for line in patch.splitlines(keepends=True):
        if line.startswith("diff --git") or (line.startswith("--- ") and (cur is None or cur[1])):
            cur = [[line], []]
            files.append(cur)
            hunk = None
        elif cur is None:
            continue
        elif line.startswith("@@"):
            m = re.match(r"@@ -(\d+)(?:,(\d+))?", line)
            hunk = [int(m.group(1)), int(m.group(2) or 1), [line]]
            cur[1].append(hunk)
        elif hunk is None:
            cur[0].append(line)
        else:
            hunk[2].append(line)
    return files


def path_of(header):
    for line in header:
        if line.startswith("--- a/"):
            return line[6:].strip()
    m = re.match(r"diff --git a/(\S+)", header[0])
    return m.group(1) if m else None


def func_spans(src):
    out = []
    try:
        tree = ast.parse(src)
    except (SyntaxError, ValueError):
        return out

    def walk(node, stack):
        for c in ast.iter_child_nodes(node):
            if isinstance(c, (ast.FunctionDef, ast.AsyncFunctionDef)):
                out.append((c.lineno, c.end_lineno or c.lineno, ".".join(stack + [c.name])))
                walk(c, stack + [c.name, "<locals>"])
            elif isinstance(c, ast.ClassDef):
                walk(c, stack + [c.name])
            else:
                walk(c, stack)
    walk(tree, [])
    return out


def hunk_functions(path, root, hunk):
    f = root / path if path else None
    spans = func_spans(f.read_text(errors="replace")) if f is not None and f.exists() else []
    old, changed = hunk[0], []
    for line in hunk[2][1:]:
        if line.startswith("-"):
            changed.append(old)
            old += 1
        elif line.startswith("+"):
            changed.append(max(old - 1, 1))
        elif not line.startswith("\\"):
            old += 1
    fns = set()
    for ln in changed:
        inside = [(e - s, q) for s, e, q in spans if s <= ln <= e]
        fns.add(f"{path}::{min(inside)[1]}" if inside else f"{path}::<module>")
    return fns


def gold_groups(patch, root):
    """-> files, {group key: [(file_idx, hunk_idx)]}, sorted function-group keys (no <module>)"""
    files = split_patch(patch)
    groups = {}
    for fi, (hdr, hunks) in enumerate(files):
        p = path_of(hdr)
        for hi, h in enumerate(hunks):
            groups.setdefault(" | ".join(sorted(hunk_functions(p, root, h))), []).append((fi, hi))
    return files, groups, sorted(k for k in groups if "<module>" not in k)


def assemble(files, keep):
    out = []
    for fi, (hdr, hunks) in enumerate(files):
        kept = [h for hi, h in enumerate(hunks) if (fi, hi) in keep]
        if kept:
            out.append("".join(hdr) + "".join("".join(h[2]) for h in kept))
    return "".join(out)


def touched_functions(patch, root):
    out = set()
    for hdr, hunks in split_patch(patch):
        p = path_of(hdr)
        for h in hunks:
            out |= {k for k in hunk_functions(p, root, h) if "<module>" not in k}
    return out


def covers(group_key, funcs):
    for g in group_key.split(" | "):
        for f in funcs:
            if f == g or f.startswith(g + ".") or g.startswith(f + "."):
                return True
    return False


# ------------------------------------------------------------------ base workspaces + patch combination
def materialise(inp, t):
    dst = WS / t["instance_id"]
    if dst.exists():
        return dst
    import swegemma.sandbox.subprocess as sp
    from swegemma.deduplication import resolve_task_snapshot_paths
    from swegemma.harness.container_setup import extract_snapshot
    mgr = sp.SubprocessManager(timeout_seconds=900)
    snap, base, patch = resolve_task_snapshot_paths(inp.snapshots, t["instance_id"], t["repo"])
    sid = mgr.start()
    try:
        extract_snapshot(mgr, sid, snap, patch_path=patch, base_snapshot_path=base)
        ws = pathlib.Path(mgr._sandboxes[sid]["workspace"])
        shutil.copytree(ws, dst, ignore=shutil.ignore_patterns(".git"), symlinks=True)
    finally:
        mgr.stop(sid)
    return dst


def git(cwd, *args, inp=None):
    return subprocess.run(["git", "-c", "user.email=a@a", "-c", "user.name=a", *args], cwd=cwd, input=inp,
                          capture_output=True, text=True)


def combine(root, patches):
    """Apply patches in order to a copy of the touched base files; -> (combined diff, how) or (None, why)."""
    text = "".join(patches)
    paths = set(re.findall(r"^(?:---|\+\+\+) [ab]/(\S+)", text, re.M))
    tmp = pathlib.Path(tempfile.mkdtemp())
    try:
        for p in paths:
            src = root / p
            if src.is_file():
                (tmp / p).parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(src, tmp / p)
        git(tmp, "init", "-q")
        git(tmp, "add", "-A")
        git(tmp, "commit", "-qm", "base", "--allow-empty")
        how = []
        for pt in patches:
            if not pt.strip():
                continue
            pt = pt if pt.endswith("\n") else pt + "\n"
            r = git(tmp, "apply", "--whitespace=nowarn", "-", inp=pt)
            if r.returncode:
                r = subprocess.run(["patch", "-p1", "-F3", "--no-backup-if-mismatch", "-s"], cwd=tmp, input=pt,
                                   capture_output=True, text=True)
                if r.returncode:
                    return None, "apply failed"
                how.append("patch-fuzz")
            else:
                how.append("git-apply")
        git(tmp, "add", "-A")
        d = git(tmp, "diff", "--cached", "HEAD")
        return d.stdout, "+".join(how)
    finally:
        shutil.rmtree(tmp, ignore_errors=True)


# ------------------------------------------------------------------ main
def runs_from_payload():
    with tarfile.open(PAYLOAD) as tf:
        tf.extractall(IN, filter="data")
    runs = {}
    for pt in sorted(IN.rglob("per_task.jsonl")):
        s = str(pt)
        name = ("v14_new60" if "v14" in s else "v7_new60") if "new60" in s else \
            "v7_old54_" + re.search(r"rep\d", s).group(0)
        pdir = pt.parent.parent / "merged" / "patches"
        rows = [json.loads(l) for l in open(pt)]
        runs[name] = {r["instance_id"]: (r, (pdir / f"{r['instance_id']}.patch").read_text(errors="replace")
                                         if (pdir / f"{r['instance_id']}.patch").exists() else "") for r in rows}
    old_loo = {}
    if (IN / "xg" / "results.jsonl").exists():
        meta = {json.loads(l)["id"]: json.loads(l) for l in open(IN / "xg" / "tasks_meta.jsonl")}
        res = [json.loads(l) for l in open(IN / "xg" / "results.jsonl")]
        for iid, m in meta.items():
            g = m.get("gold_function_groups") or []
            loo = {r["mode"]: r["resolved"] for r in res if r["id"] == iid and r["mode"].startswith("loo")}
            old_loo[iid] = {k: (loo.get(f"loo{i}") is False) for i, k in enumerate(g)} if len(g) >= 2 else \
                {k: True for k in g}
    return runs, old_loo


def main():
    t0 = time.time()
    s = Settings.load()
    set_scorer_env()
    patch_sandbox_venv()
    inp = Inputs.locate(False, need_model=False)
    use_scorer_sandbox(inp.data)
    make_sandbox_offline()
    OUT.mkdir(parents=True, exist_ok=True)
    gold = gold_patches(inp)
    tasks = {t["instance_id"]: t for t in inp.load_tasks()}
    valid = set(json.loads(VALID_TASKS.read_text()))
    runs, old_loo = runs_from_payload()
    new60 = sorted({i for r in ("v7_new60", "v14_new60") for i in runs.get(r, {})} & valid)
    if s.max_tasks:  # smoke run: a few new-60 tasks and only the agent runs on them
        new60 = new60[: s.max_tasks]
        runs = {k: {i: v for i, v in r.items() if i in new60} for k, r in runs.items()}
    involved = sorted(({i for r in runs.values() for i in r} | set(new60)) & valid)
    log(f"runs {sorted(runs)}; new-60 valid {len(new60)}; tasks involved {len(involved)}")

    groups = {}
    for iid in involved:
        try:
            root = materialise(inp, tasks[iid])
            files, gmap, fg = gold_groups(gold[iid], root)
            groups[iid] = (root, files, gmap, fg)
        except Exception as e:  # noqa: BLE001
            log("materialise/groups failed", iid, type(e).__name__, str(e)[:200])
    (OUT / "groups.json").write_text(json.dumps({i: v[3] for i, v in groups.items()}, indent=1))
    log(f"groups built for {len(groups)} tasks in {(time.time() - t0) / 60:.1f} min")

    rows = []

    def run(label, patches, extra):
        if not patches:
            return {}
        if s.hours_left() < 0.3:
            log("budget: skipping", label)
            return {}
        t = time.time()
        r = verify(inp, patches, label)
        for iid, v in r.items():
            rows.append({"task": iid, "mode": label, **extra.get(iid, {}), "resolved": v["resolved"], "exit": v["exit"],
                         "error": v["error"]})
        with open(OUT / "results.jsonl", "w") as f:
            f.write("".join(json.dumps(x) + "\n" for x in rows))
        log(f"{label}: {sum(v['resolved'] for v in r.values())}/{len(r)} resolved in {(time.time() - t) / 60:.1f} min")
        return r

    # ---------------- A3: necessity on new-60
    multi = [i for i in new60 if i in groups and len(groups[i][3]) >= 2]
    run("gold", {i: gold[i] for i in multi}, {})
    run("empty", {i: "" for i in multi}, {})
    nec = {}
    for k in range(max((len(groups[i][3]) for i in multi), default=0)):
        pats = {}
        for i in multi:
            _, files, gmap, fg = groups[i]
            if k < len(fg):
                drop = set(gmap[fg[k]])
                pats[i] = assemble(files, {(fi, hi) for hunks in gmap.values() for fi, hi in hunks} - drop)
        r = run(f"loo{k}", pats, {})
        for i, v in r.items():
            nec.setdefault(i, {})[groups[i][3][k]] = not v["resolved"]
    for i in new60:
        if i in groups and len(groups[i][3]) == 1:
            nec[i] = {groups[i][3][0]: True}
    (OUT / "necessity_new60.json").write_text(json.dumps(nec, indent=1))

    # ---------------- A4: completion counterfactual
    cases = []
    for rname, tasks_r in runs.items():
        for iid, (r, apatch) in tasks_r.items():
            if iid not in groups or r.get("resolved") or not apatch.strip():
                continue
            root, files, gmap, fg = groups[iid]
            if len(fg) < 1:
                continue
            tf = touched_functions(apatch, root)
            covered = [g for g in fg if covers(g, tf)]
            missed = [g for g in fg if g not in covered]
            if not covered or not missed:
                continue
            module = {h for k, hs in gmap.items() if "<module>" in k for h in hs}
            keep = {h for g in missed for h in gmap[g]} | module
            missed_patch = assemble(files, keep)
            comp, how = combine(root, [apatch, missed_patch])
            if comp is None:  # retry without the module-level hunks
                missed_patch = assemble(files, {h for g in missed for h in gmap[g]})
                comp, how = combine(root, [apatch, missed_patch])
                how = (how or "") + "|no-module"
            nlab = nec.get(iid) or old_loo.get(iid) or {}
            cases.append({"run": rname, "task": iid, "covered": covered, "missed": missed,
                          "missed_necessary": [g for g in missed if nlab.get(g)], "necessity_known": bool(nlab),
                          "apply": how if comp is not None else "apply failed", "_agent": apatch,
                          "_comp": comp, "_missed": missed_patch})
    log(f"A4 cases: {len(cases)} ({sum(c['_comp'] is not None for c in cases)} combinable)")
    for rname in sorted({c["run"] for c in cases}):
        cs = [c for c in cases if c["run"] == rname]
        ex = {c["task"]: {"run": rname} for c in cs}
        run(f"agent:{rname}", {c["task"]: c["_agent"] for c in cs}, ex)
        run(f"comp:{rname}", {c["task"]: c["_comp"] for c in cs if c["_comp"] is not None}, ex)
        run(f"missed:{rname}", {c["task"]: c["_missed"] for c in cs}, ex)
    (OUT / "cases.json").write_text(json.dumps([{k: v for k, v in c.items() if not k.startswith("_")} for c in cases],
                                               indent=1))

    # ---------------- report (pre-registered outcomes)
    res = {(x["mode"], x["task"]): x["resolved"] for x in rows}
    for c in cases:
        c["agent_ok"] = res.get((f"agent:{c['run']}", c["task"]))
        c["comp_ok"] = res.get((f"comp:{c['run']}", c["task"]))
        c["missed_ok"] = res.get((f"missed:{c['run']}", c["task"]))
        c["pure"] = bool(c["comp_ok"]) and c["missed_ok"] is False

    def summ(cs):
        ok = [c for c in cs if c["comp_ok"] is not None and c["missed_ok"] is not None]
        n = len(ok)
        return {"cases": len(cs), "evaluated": n, "apply_failed": sum(c["_comp"] is None for c in cs),
                "agent_alone_resolved": sum(bool(c["agent_ok"]) for c in cs),
                "completion_resolved": sum(bool(c["comp_ok"]) for c in ok),
                "missed_only_resolved": sum(bool(c["missed_ok"]) for c in ok),
                "pure_incompleteness": sum(c["pure"] for c in ok),
                "completion_rate": round(sum(bool(c["comp_ok"]) for c in ok) / n, 3) if n else None,
                "pure_rate": round(sum(c["pure"] for c in ok) / n, 3) if n else None}

    multi_nec = sorted(i for i, d in nec.items() if sum(d.values()) >= 2)
    report = {
        "A3": {"new60_valid": len(new60), "multi_group": len(multi), "with_necessity_labels": len(nec),
               "tasks_with_>=2_necessary_groups": len(multi_nec), "ids": multi_nec,
               "gold_resolved": sum(bool(res.get(("gold", i))) for i in multi),
               "empty_resolved": sum(bool(res.get(("empty", i))) for i in multi)},
        "A4": {"pooled": summ(cases),
               "missed_a_necessary_group": summ([c for c in cases if c["missed_necessary"]]),
               **{r: summ([c for c in cases if c["run"] == r]) for r in sorted({c["run"] for c in cases})}},
        "hours": round((time.time() - t0) / 3600, 2)}
    (WORK / "necessity_report.json").write_text(json.dumps(report, indent=1))
    (OUT / "cases.json").write_text(json.dumps([{k: v for k, v in c.items() if not k.startswith("_")} for c in cases],
                                               indent=1))
    log("A3:", json.dumps(report["A3"]))
    log("A4 pooled:", json.dumps(report["A4"]["pooled"]))
    log("A4 missed a necessary group:", json.dumps(report["A4"]["missed_a_necessary_group"]))


if __name__ == "__main__":
    main()


# ================================================================== A12 site-swap (+ A5 typology), pre-registered Oct 24
# For every unresolved agent patch that edits >= 1 gold function group g: start from the gold-patched tree and replace
# g's function(s) with the agent-patched version (AST source segment, function level). Classify before testing:
# identical / conflict / coupled / clean; then run the fail-to-pass tests on clean + coupled swaps.
SWAP_OUT = WORK / "siteswap"
NORM_WS = re.compile(r"\s+")


def patched_texts(root, patches):
    """{path: text or None (deleted)} for every file the patches touch, after applying them to the base; or None."""
    text = "".join(patches)
    paths = set(re.findall(r"^(?:---|\+\+\+) [ab]/(\S+)", text, re.M))
    tmp = pathlib.Path(tempfile.mkdtemp())
    try:
        for p in paths:
            src = root / p
            if src.is_file():
                (tmp / p).parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(src, tmp / p)
        git(tmp, "init", "-q")
        for pt in patches:
            if not pt.strip():
                continue
            pt = pt if pt.endswith("\n") else pt + "\n"
            if git(tmp, "apply", "--whitespace=nowarn", "-", inp=pt).returncode and subprocess.run(
                    ["patch", "-p1", "-F3", "--no-backup-if-mismatch", "-s"], cwd=tmp, input=pt,
                    capture_output=True, text=True).returncode:
                return None
        return {p: ((tmp / p).read_text(errors="replace") if (tmp / p).is_file() else None) for p in paths}
    finally:
        shutil.rmtree(tmp, ignore_errors=True)


def diff_from_texts(root, texts):
    """Unified diff base -> texts (path -> new text or None)."""
    tmp = pathlib.Path(tempfile.mkdtemp())
    try:
        for p in texts:
            if (root / p).is_file():
                (tmp / p).parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(root / p, tmp / p)
        git(tmp, "init", "-q")
        git(tmp, "add", "-A")
        git(tmp, "commit", "-qm", "base", "--allow-empty")
        for p, t in texts.items():
            f = tmp / p
            if t is None:
                f.unlink(missing_ok=True)
            else:
                f.parent.mkdir(parents=True, exist_ok=True)
                f.write_text(t)
        git(tmp, "add", "-A")
        return git(tmp, "diff", "--cached", "HEAD").stdout
    finally:
        shutil.rmtree(tmp, ignore_errors=True)


def span_of(text, qual):
    for s_, e_, q in func_spans(text or ""):
        if q == qual:
            return s_, e_
    return None


def defined_names(text, skip=()):
    """Names a file defines OUTSIDE the line spans in `skip`: module- and class-level defs, imports and assignments,
    plus attributes assigned on self/cls anywhere (a field set in __init__ couples the methods that read it)."""
    out = set()
    try:
        tree = ast.parse(text or "")
    except (SyntaxError, ValueError):
        return out

    def inside(n):
        return any(a <= getattr(n, "lineno", -1) <= b for a, b in skip)

    def scope(body):
        for n in body:
            if inside(n):
                continue
            if isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef)):
                out.add(n.name)
                if isinstance(n, ast.ClassDef):
                    scope(n.body)
            elif isinstance(n, (ast.Import, ast.ImportFrom)):
                out.update((a.asname or a.name).split(".")[0] for a in n.names)
            elif isinstance(n, (ast.Assign, ast.AnnAssign, ast.AugAssign)):
                for t in (n.targets if isinstance(n, ast.Assign) else [n.target]):
                    out.update(x.id for x in ast.walk(t) if isinstance(x, ast.Name))
    scope(tree.body)
    for n in ast.walk(tree):
        if isinstance(n, ast.Attribute) and isinstance(n.ctx, ast.Store) and isinstance(n.value, ast.Name) \
                and n.value.id in ("self", "cls") and not inside(n):
            out.add(n.attr)
    return out


def referenced_names(segment):
    import textwrap
    try:
        tree = ast.parse(textwrap.dedent(segment))
    except (SyntaxError, ValueError):
        return set()
    out = set()
    for n in ast.walk(tree):
        if isinstance(n, ast.Name) and isinstance(n.ctx, ast.Load):
            out.add(n.id)
        elif isinstance(n, ast.Attribute) and isinstance(n.value, ast.Name) and n.value.id in ("self", "cls"):
            out.add(n.attr)
    return out


def build_swap(root, gold_t, agent_t, base_t, group_key):
    """-> (class, patch or None, detail)"""
    funcs = [f for f in group_key.split(" | ")]
    funcs = [f for f in funcs if not any(f != o and f.startswith(o + ".") for o in funcs)]  # outermost only
    texts = dict(gold_t)
    same = True
    segs = []
    by_path = {}
    for f in funcs:
        path, qual = f.split("::", 1)
        by_path.setdefault(path, []).append(qual)
    for path, quals in by_path.items():
        g_text, a_text = gold_t.get(path), agent_t.get(path)
        if g_text is None or a_text is None:
            return "conflict", None, f"file missing in {'gold' if g_text is None else 'agent'} tree: {path}"
        g_lines, a_lines = g_text.splitlines(keepends=True), a_text.splitlines(keepends=True)
        repl = []
        for q in quals:
            gs, asp = span_of(g_text, q), span_of(a_text, q)
            if not gs or not asp:
                return "conflict", None, f"function {q} missing in {'gold' if not gs else 'agent'} tree"
            gseg, aseg = "".join(g_lines[gs[0] - 1:gs[1]]), "".join(a_lines[asp[0] - 1:asp[1]])
            same &= NORM_WS.sub("", gseg) == NORM_WS.sub("", aseg)
            repl.append((gs, a_lines[asp[0] - 1:asp[1]]))
            segs.append(aseg)
        for (s0, e0), new in sorted(repl, key=lambda x: -x[0][0]):
            g_lines[s0 - 1:e0] = new
        texts[path] = "".join(g_lines)
        try:
            compile(texts[path], path, "exec")
        except SyntaxError as e:
            return "conflict", None, f"py_compile: {e.msg} line {e.lineno}"
    if same:
        return "identical", None, ""
    agent_new = set()
    for p, t in agent_t.items():  # defined by the agent's edits OUTSIDE g
        skip = [span_of(t, q) for q in by_path.get(p, [])]
        agent_new |= defined_names(t, [x for x in skip if x]) - defined_names(base_t.get(p))
    gold_def = set().union(*(defined_names(t) for t in gold_t.values() if t)) if gold_t else set()
    refs = set().union(*(referenced_names(s_) for s_ in segs)) if segs else set()
    missing = sorted((refs & agent_new) - gold_def)
    patch = diff_from_texts(root, texts)
    return ("coupled" if missing else "clean"), patch, ",".join(missing[:5])


def a5_typology(groups, nec, gold, payload_dir):
    """Descriptive: oblig.py v2 relations between test-necessary groups of new-60 tasks with >= 2 of them."""
    import sys
    sys.path.insert(0, str(payload_dir / "tools"))
    import oblig
    rows, diffs = [], []
    for iid, labs in sorted(nec.items()):
        necs = [g for g, v in labs.items() if v]
        if len(necs) < 2 or iid not in groups:
            continue
        root, files, gmap, _ = groups[iid]
        idx = oblig.Index(str(root)).build()

        def keys(g):
            return [f.replace(".<locals>", "") for f in g.split(" | ")]

        def edited_idents(g):
            toks = set()
            for fi, hi in gmap.get(g, []):
                for line in files[fi][1][hi][2][1:]:
                    if line[:1] in "+-":
                        toks |= set(re.findall(r"[A-Za-z_][A-Za-z0-9_]{2,}", line[1:]))
            return toks
        for a in necs:
            for b in necs:
                if a == b:
                    continue
                ka, kb = keys(a), keys(b)
                direct = sorted({t for x in ka if x in idx.defs for y, t, _ in idx.neighbours(x)
                                 if any(y == k or y.startswith(k + ".") or k.startswith(y + ".") for k in kb)})
                two = False
                for x in ka:
                    if x not in idx.defs:
                        continue
                    for y, _, _ in idx.neighbours(x):
                        for z, _, _ in idx.neighbours(y):
                            if any(z == k or z.startswith(k + ".") for k in kb):
                                two = True
                ca = {idx.defs[x].cls for x in ka if x in idx.defs}
                cb = {idx.defs[x].cls for x in kb if x in idx.defs}
                rows.append({"task": iid, "a": a, "b": b, "direct_edge_types": direct, "two_hop": two,
                             "same_file": ka[0].split("::")[0] == kb[0].split("::")[0],
                             "same_class": bool(ca & cb - {None}),
                             "shared_edited_identifiers": sorted(edited_idents(a) & edited_idents(b))[:8]})
        diffs.append(f"## {iid}\nnecessary groups: {necs}\n")
        for g in necs:
            body = "".join("".join(files[fi][1][hi][2]) for fi, hi in gmap.get(g, []))
            diffs.append(f"### {g}\n```diff\n{''.join(body.splitlines(keepends=True)[:40])}```\n")
    return rows, "\n".join(diffs)


def main_siteswap():
    import random as _r
    t0 = time.time()
    s = Settings.load()
    set_scorer_env()
    patch_sandbox_venv()
    inp = Inputs.locate(False, need_model=False)
    use_scorer_sandbox(inp.data)
    make_sandbox_offline()
    SWAP_OUT.mkdir(parents=True, exist_ok=True)
    gold = gold_patches(inp)
    tasks = {t["instance_id"]: t for t in inp.load_tasks()}
    valid = set(json.loads(VALID_TASKS.read_text()))
    runs, old_loo = runs_from_payload()
    nec_new = json.loads((IN / "a3" / "necessity_new60.json").read_text())
    cand = sorted({i for r in runs.values() for i, (row, p) in r.items()
                   if not row.get("resolved") and p.strip()} & valid)
    if s.max_tasks:
        cand = cand[: s.max_tasks]
    groups = {}
    for iid in sorted(set(cand) | {i for i, d in nec_new.items() if sum(d.values()) >= 2} & valid):
        try:
            root = materialise(inp, tasks[iid])
            groups[iid] = (root,) + gold_groups(gold[iid], root)
        except Exception as e:  # noqa: BLE001
            log("materialise/groups failed", iid, type(e).__name__, str(e)[:200])
    log(f"site-swap: {len(cand)} candidate tasks, groups for {len(groups)} in {(time.time() - t0) / 60:.1f} min")

    def necessary(iid, g, fg):
        lab = nec_new.get(iid) or old_loo.get(iid)
        if lab and g in lab:
            return bool(lab[g])
        return True if len(fg) == 1 else None

    swaps, gold_cache = [], {}
    for rname, rr in runs.items():
        for iid, (row, apatch) in rr.items():
            if iid not in cand or iid not in groups or row.get("resolved") or not apatch.strip():
                continue
            root, files, gmap, fg = groups[iid]
            covered = [g for g in fg if covers(g, touched_functions(apatch, root))]
            if not covered:
                continue
            if iid not in gold_cache:
                gold_cache[iid] = patched_texts(root, [gold[iid]])
            agent_t = patched_texts(root, [apatch])
            gold_t = gold_cache[iid]
            for g in covered:
                if gold_t is None or agent_t is None:
                    cls_, patch, det = "conflict", None, "patch does not apply"
                else:
                    paths = {f.split("::")[0] for f in g.split(" | ")} | set(agent_t)
                    base_t = {p: ((root / p).read_text(errors="replace") if (root / p).is_file() else None) for p in paths}
                    full_gold = {**{p: base_t.get(p) for p in paths}, **gold_t}
                    full_agent = {**{p: base_t.get(p) for p in paths}, **agent_t}
                    try:
                        cls_, patch, det = build_swap(root, full_gold, full_agent, base_t, g)
                    except Exception as e:  # noqa: BLE001
                        cls_, patch, det = "conflict", None, f"{type(e).__name__}: {str(e)[:120]}"
                swaps.append({"run": rname, "task": iid, "group": g, "necessary": necessary(iid, g, fg),
                              "class": cls_, "detail": det, "_patch": patch})
    log(f"swaps: {len(swaps)}; classes {dict((c, sum(x['class'] == c for x in swaps)) for c in ('identical', 'conflict', 'coupled', 'clean'))}")

    rows = []

    def run(label, patches, meta):
        if not patches or s.hours_left() < 0.3:
            return {}
        t = time.time()
        r = verify(inp, patches, label)
        for iid, v in r.items():
            rows.append({"task": iid, "mode": label, **meta.get(iid, {}), "resolved": v["resolved"], "exit": v["exit"],
                         "error": v["error"]})
        (SWAP_OUT / "results.jsonl").write_text("".join(json.dumps(x) + "\n" for x in rows))
        log(f"{label}: {sum(v['resolved'] for v in r.values())}/{len(r)} resolved in {(time.time() - t) / 60:.1f} min")
        return r

    gold_ok = run("gold_rerun", {i: gold[i] for i in sorted({x["task"] for x in swaps})}, {})
    todo = [x for x in swaps if x["class"] in ("clean", "coupled") and gold_ok.get(x["task"], {}).get("resolved")]
    k = 0
    while todo:  # one patch per task per Evaluator call
        batch, rest = {}, []
        for x in todo:
            if x["task"] in batch:
                rest.append(x)
            else:
                batch[x["task"]] = x
        r = run(f"swap{k}", {t: x["_patch"] for t, x in batch.items()}, {})
        for t, x in batch.items():
            x["resolved"] = r.get(t, {}).get("resolved")
        if not r:
            break
        todo, k = rest, k + 1

    for x in swaps:
        x["gold_ok"] = bool(gold_ok.get(x["task"], {}).get("resolved"))
        x["wrong"] = (x.get("resolved") is False) if x["class"] in ("clean", "coupled") else None
    (SWAP_OUT / "swaps.json").write_text(json.dumps([{k_: v for k_, v in x.items() if k_ != "_patch"} for x in swaps],
                                                    indent=1))

    def rate(xs):
        xs = [x for x in xs if x["wrong"] is not None and x["gold_ok"]]
        return (sum(x["wrong"] for x in xs), len(xs))

    prim = [x for x in swaps if x["class"] == "clean" and x["necessary"]]
    w, n = rate(prim)
    by_task = {}
    for x in prim:
        if x["wrong"] is not None and x["gold_ok"]:
            by_task.setdefault(x["task"], []).append(x["wrong"])
    boots = []
    ts = list(by_task)
    rnd = _r.Random(0)
    for _ in range(10000):
        smp = [rnd.choice(ts) for _ in ts] if ts else []
        vals = [v for t in smp for v in by_task[t]]
        boots.append(sum(vals) / len(vals) if vals else 0)
    boots.sort()
    cls_counts = {c: sum(x["class"] == c for x in swaps) for c in ("identical", "conflict", "coupled", "clean")}
    report = {
        "swaps": len(swaps), "classes": cls_counts,
        "conflict_plus_coupled_share": round((cls_counts["conflict"] + cls_counts["coupled"]) / max(1, len(swaps)), 3),
        "primary_clean_necessary": {"wrong": w, "n": n, "rate": round(w / n, 3) if n else None,
                                    "ci95_task_bootstrap": [round(boots[250], 3), round(boots[9750], 3)] if ts else None},
        "task_level": {"tasks": len(by_task), "tasks_with_a_wrong_clean_necessary_swap": sum(any(v) for v in by_task.values())},
        "all_clean": dict(zip(("wrong", "n"), rate([x for x in swaps if x["class"] == "clean"]))),
        "coupled": dict(zip(("wrong", "n"), rate([x for x in swaps if x["class"] == "coupled"]))),
        "gold_rerun_failed_tasks": sorted(t for t, v in gold_ok.items() if not v.get("resolved")),
    }
    try:
        rows5, diffs = a5_typology({i: g for i, g in groups.items()}, nec_new, gold, IN)
        (SWAP_OUT / "a5_pairs.json").write_text(json.dumps(rows5, indent=1))
        (SWAP_OUT / "a5_diffs.md").write_text(diffs)
        report["A5"] = {"tasks": len({r_["task"] for r_ in rows5}), "pairs": len(rows5),
                        "same_file": sum(r_["same_file"] for r_ in rows5), "same_class": sum(r_["same_class"] for r_ in rows5),
                        "direct_any_edge": sum(bool(r_["direct_edge_types"]) for r_ in rows5),
                        "direct_call_edge": sum(any(t in ("CALLER", "CALLEE", "FWD", "FWDDOWN", "INST") for t in r_["direct_edge_types"]) for r_ in rows5),
                        "two_hop": sum(r_["two_hop"] for r_ in rows5),
                        "shared_edited_identifier": sum(bool(r_["shared_edited_identifiers"]) for r_ in rows5)}
        per = {}
        for r_ in rows5:
            per.setdefault(r_["task"], []).append(r_)
        macro = {}
        for key in ("same_file", "same_class", "two_hop"):
            macro[key] = round(sum(sum(bool(r_[key]) for r_ in v) / len(v) for v in per.values()) / len(per), 3)
        macro["direct_any_edge"] = round(sum(sum(bool(r_["direct_edge_types"]) for r_ in v) / len(v) for v in per.values()) / len(per), 3)
        macro["shared_edited_identifier"] = round(sum(sum(bool(r_["shared_edited_identifiers"]) for r_ in v) / len(v) for v in per.values()) / len(per), 3)
        report["A5"]["task_macro"] = macro
        report["A5"]["pairs_per_task"] = {t: len(v) for t, v in sorted(per.items(), key=lambda kv: -len(kv[1]))}
    except Exception as e:  # noqa: BLE001
        report["A5_error"] = f"{type(e).__name__}: {e}"
    report["hours"] = round((time.time() - t0) / 3600, 2)
    (WORK / "siteswap_report.json").write_text(json.dumps(report, indent=1))
    log("A12:", json.dumps({k_: report[k_] for k_ in ("classes", "conflict_plus_coupled_share", "primary_clean_necessary", "task_level")}))
    log("A5:", json.dumps(report.get("A5") or report.get("A5_error")))


# ================================================================== A12v validated site-swap, pre-registered Oct 27
# Fixes the A12 construct problems found in Atul's audit: decorators are part of the function span; names the agent's
# version of g loads must resolve in the SAME file of the gold tree; attributes the agent added elsewhere in the class
# are flagged; signature changes are their own class. Identity controls (gold->gold, agent->agent), every hybrid patch
# and 4,000 chars of test output are kept, and failures are typed from the output by a rule fixed in the registration.
SWAP2_OUT = WORK / "siteswap2"
CLASS_ORDER = ["conflict", "namespace-incompatible", "attribute-incompatible", "interface-changed", "identical",
               "unknown", "validated-clean"]


def func_nodes(text):
    """{qual: (first line incl. decorators, last line, node)} with the exec-graph qualname rule."""
    out = {}
    try:
        tree = ast.parse(text or "")
    except (SyntaxError, ValueError):
        return out

    def walk(node, stack):
        for c in ast.iter_child_nodes(node):
            if isinstance(c, (ast.FunctionDef, ast.AsyncFunctionDef)):
                q = ".".join(stack + [c.name])
                start = min([c.lineno] + [d.lineno for d in c.decorator_list])
                out[q] = (start, c.end_lineno or c.lineno, c)
                walk(c, stack + [c.name, "<locals>"])
            elif isinstance(c, ast.ClassDef):
                walk(c, stack + [c.name])
            else:
                walk(c, stack)
    walk(tree, [])
    return out


def module_names(text):
    """Names bound at module level (descending into if/try/with/for, not into defs) and whether `import *` occurs."""
    names, star = set(), False
    try:
        tree = ast.parse(text or "")
    except (SyntaxError, ValueError):
        return names, star

    def scope(body):
        nonlocal star
        for n in body:
            if isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef)):
                names.add(n.name)
            elif isinstance(n, (ast.Import, ast.ImportFrom)):
                for a in n.names:
                    if a.name == "*":
                        star = True
                    else:
                        names.add((a.asname or a.name).split(".")[0])
            elif isinstance(n, (ast.Assign, ast.AnnAssign, ast.AugAssign)):
                for t in (n.targets if isinstance(n, ast.Assign) else [n.target]):
                    names.update(x.id for x in ast.walk(t) if isinstance(x, ast.Name))
            elif isinstance(n, (ast.For, ast.AsyncFor)):
                names.update(x.id for x in ast.walk(n.target) if isinstance(x, ast.Name))
            for field in ("body", "orelse", "finalbody", "handlers"):
                sub = getattr(n, field, None)
                if isinstance(sub, list) and not isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef, ast.ClassDef)):
                    scope([x for x in sub if isinstance(x, ast.stmt)] + [y for h in sub if isinstance(h, ast.ExceptHandler)
                                                                         for y in h.body])
            if isinstance(n, ast.With):
                for it in n.items:
                    if it.optional_vars is not None:
                        names.update(x.id for x in ast.walk(it.optional_vars) if isinstance(x, ast.Name))
    scope(tree.body)
    return names, star


def bound_in(fn):
    out = set()
    for n in ast.walk(fn):
        if isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef, ast.Lambda)):
            a = n.args
            out.update(x.arg for x in a.posonlyargs + a.args + a.kwonlyargs)
            out.update(x.arg for x in (a.vararg, a.kwarg) if x)
            if not isinstance(n, ast.Lambda):
                out.add(n.name)
        elif isinstance(n, ast.ClassDef):
            out.add(n.name)
        elif isinstance(n, ast.Name) and isinstance(n.ctx, (ast.Store, ast.Del)):
            out.add(n.id)
        elif isinstance(n, (ast.Import, ast.ImportFrom)):
            out.update((a.asname or a.name).split(".")[0] for a in n.names)
        elif isinstance(n, ast.ExceptHandler) and n.name:
            out.add(n.name)
        elif hasattr(ast, "MatchAs") and isinstance(n, (ast.MatchAs, ast.MatchStar)) and getattr(n, "name", None):
            out.add(n.name)
        elif isinstance(n, (ast.Global, ast.Nonlocal)):
            out.update(n.names)
    return out


def loaded_names(fn):
    return {n.id for n in ast.walk(fn) if isinstance(n, ast.Name) and isinstance(n.ctx, ast.Load)}


def class_attrs(text, cls_qual):
    """Attributes set on self/cls in any method of the class, plus class-level names; cls_qual like 'A' or 'A.B'."""
    out = set()
    try:
        tree = ast.parse(text or "")
    except (SyntaxError, ValueError):
        return out
    node, parts = tree, cls_qual.split(".")
    for p_ in parts:
        node = next((c for c in ast.iter_child_nodes(node) if isinstance(c, ast.ClassDef) and c.name == p_), None)
        if node is None:
            return out
    for st in node.body:
        if isinstance(st, (ast.Assign, ast.AnnAssign)):
            for t in (st.targets if isinstance(st, ast.Assign) else [st.target]):
                out.update(x.id for x in ast.walk(t) if isinstance(x, ast.Name))
        elif isinstance(st, (ast.FunctionDef, ast.AsyncFunctionDef)):
            out.add(st.name)
    for n in ast.walk(node):
        if isinstance(n, ast.Attribute) and isinstance(n.ctx, ast.Store) and isinstance(n.value, ast.Name) \
                and n.value.id in ("self", "cls"):
            out.add(n.attr)
    return out


def signature(fn):
    a = fn.args
    return ([x.arg for x in a.posonlyargs], [x.arg for x in a.args], bool(a.vararg), [x.arg for x in a.kwonlyargs],
            bool(a.kwarg), len(a.defaults), sum(d is not None for d in a.kw_defaults))


def build_swap_v2(gold_t, agent_t, group_key):
    """-> (class, swapped texts or None, detail). Classes as in CLASS_ORDER (worst wins across the group's functions)."""
    import builtins
    BUILTIN = set(dir(builtins)) | {"__name__", "__file__", "__doc__", "__class__", "__spec__", "__package__"}
    funcs = group_key.split(" | ")
    funcs = [f for f in funcs if not any(f != o and f.startswith(o + ".") for o in funcs)]
    by_path = {}
    for f in funcs:
        path, qual = f.split("::", 1)
        by_path.setdefault(path, []).append(qual)
    texts, worst, details, all_same = dict(gold_t), "validated-clean", [], True

    def bump(c, d):
        nonlocal worst
        if CLASS_ORDER.index(c) < CLASS_ORDER.index(worst):
            worst = c
        details.append(d)
    for path, quals in by_path.items():
        g_text, a_text = gold_t.get(path), agent_t.get(path)
        if g_text is None or a_text is None:
            return "conflict", None, f"file missing in {'gold' if g_text is None else 'agent'} tree: {path}"
        gn, an = func_nodes(g_text), func_nodes(a_text)
        g_lines, a_lines = g_text.splitlines(keepends=True), a_text.splitlines(keepends=True)
        mod, star = module_names(g_text)
        repl = []
        for q in quals:
            if q not in gn or q not in an:
                return "conflict", None, f"function {q} missing in {'gold' if q not in gn else 'agent'} tree"
            gs, ge, gnode = gn[q]
            as_, ae, anode = an[q]
            same = ast.dump(gnode, include_attributes=False) == ast.dump(anode, include_attributes=False)
            all_same &= same
            repl.append(((gs, ge), a_lines[as_ - 1:ae]))
            if same:
                continue
            missing = sorted(loaded_names(anode) - bound_in(anode) - BUILTIN - mod)
            if missing:
                if star:
                    bump("unknown", f"{q}: star import; unresolved {missing[:4]}")
                else:
                    bump("namespace-incompatible", f"{q}: unresolved in gold {path}: {missing[:4]}")
            cls_q = q.rsplit(".", 1)[0] if "." in q and "<locals>" not in q else None
            if cls_q:
                added = class_attrs(a_text, cls_q) - class_attrs(g_text, cls_q)
                reads = {n.attr for n in ast.walk(anode) if isinstance(n, ast.Attribute) and isinstance(n.ctx, ast.Load)
                         and isinstance(n.value, ast.Name) and n.value.id in ("self", "cls")}
                # attributes the agent set only inside g itself are fine
                inside = {n.attr for n in ast.walk(anode) if isinstance(n, ast.Attribute) and isinstance(n.ctx, ast.Store)}
                bad = sorted((reads & added) - inside)
                if bad:
                    bump("attribute-incompatible", f"{q}: reads agent-added attrs {bad[:4]}")
            if signature(anode) != signature(gnode):
                bump("interface-changed", f"{q}: signature differs")
        for (s0, e0), new in sorted(repl, key=lambda x: -x[0][0]):
            g_lines[s0 - 1:e0] = new
        texts[path] = "".join(g_lines)
        try:
            compile(texts[path], path, "exec")
        except SyntaxError as e:
            return "conflict", None, f"py_compile: {e.msg} line {e.lineno}"
    if all_same:
        return "identical", texts, ""
    return worst, texts, "; ".join(details)[:300]


def identity_swap(t_texts, group_key):
    """Replace each function of the group with its own source in the same tree; returns True if the text is unchanged."""
    for f in group_key.split(" | "):
        path, qual = f.split("::", 1)
        t = t_texts.get(path)
        if t is None:
            return False
        nodes = func_nodes(t)
        if qual not in nodes:
            return False
        s0, e0, _ = nodes[qual]
        lines = t.splitlines(keepends=True)
        if "".join(lines[:s0 - 1] + lines[s0 - 1:e0] + lines[e0:]) != t:
            return False
    return True


def failure_type(row):
    if row.get("resolved"):
        return "pass"
    out = (row.get("tail") or "") + " " + str(row.get("error") or "")
    if row.get("exit") == 124 or "timed out" in out.lower():
        return "timeout"
    comp = re.search(r"NameError|ImportError|ModuleNotFoundError|SyntaxError|error collecting|errors during collection", out)
    assertion = re.search(r"AssertionError|\bassert\b|^FAILED ", out, re.M)
    if comp and not assertion:
        return "composition"
    if re.search(r"^FAILED |\d+ failed", out, re.M):
        return "behavioural"
    return "other"


def verify_full(inp, patches, label):
    """As preflight.verify, but keeps 4,000 chars of test output."""
    from adk_eval_core.tracing import SessionTrace
    from adk_submission import ModelRegistry
    from swegemma.config import EvalConfig
    from swegemma.evaluate import Evaluator
    import asyncio

    async def fake_agent(self, task, snapshot_path, task_index, total_tasks, **kw):
        return patches[task.instance_id], None, SessionTrace()

    Evaluator._run_agent_sandbox = fake_agent
    dummy = WORK / "preflight_submission"
    dummy.mkdir(exist_ok=True)
    cfg = EvalConfig(tasks_path=inp.tasks_path, snapshots_dir=inp.snapshots, results_dir=WORK / f"v2_{label}",
                     submission_dir=dummy, models=ModelRegistry(), sandbox="subprocess", wheels_dir=inp.data / "wheels",
                     graph_dir=str(inp.data / "graphs"), embeddings_dir=str(inp.data / "embeddings"),
                     task_ids=sorted(patches), concurrency=4, display_mode="quiet", max_time_minutes=30,
                     timeout_seconds=600)
    res = asyncio.run(Evaluator(cfg).run())
    return {r.instance_id: {"resolved": bool(r.resolved), "exit": r.test_exit_code, "error": r.error,
                            "tail": (r.test_output or "")[-4000:]} for r in res.task_results}


def main_siteswap_v2():
    import keyword
    import random as _r
    t0 = time.time()
    s = Settings.load()
    set_scorer_env()
    patch_sandbox_venv()
    inp = Inputs.locate(False, need_model=False)
    use_scorer_sandbox(inp.data)
    make_sandbox_offline()
    (SWAP2_OUT / "patches").mkdir(parents=True, exist_ok=True)
    gold = gold_patches(inp)
    tasks = {t["instance_id"]: t for t in inp.load_tasks()}
    valid = set(json.loads(VALID_TASKS.read_text()))
    runs, old_loo = runs_from_payload()
    nec_new = json.loads((IN / "a3" / "necessity_new60.json").read_text())
    v1 = {(x["run"], x["task"], x["group"]): x for x in json.loads((IN / "a12" / "swaps.json").read_text())}
    cand = sorted({t for (_, t, _) in v1} & valid)
    if s.max_tasks:
        cand = cand[: s.max_tasks]
    groups = {}
    for iid in sorted(set(cand) | {i for i, d in nec_new.items() if sum(d.values()) >= 2} & valid):
        try:
            root = materialise(inp, tasks[iid])
            groups[iid] = (root,) + gold_groups(gold[iid], root)
        except Exception as e:  # noqa: BLE001
            log("materialise/groups failed", iid, type(e).__name__, str(e)[:200])
    log(f"A12v: {len(cand)} tasks, groups for {len(groups)} in {(time.time() - t0) / 60:.1f} min")

    swaps, gold_cache = [], {}
    for (rname, iid, g), x1 in sorted(v1.items()):
        if iid not in cand or iid not in groups:
            continue
        root = groups[iid][0]
        apatch = runs.get(rname, {}).get(iid, (None, ""))[1]
        if iid not in gold_cache:
            gold_cache[iid] = patched_texts(root, [gold[iid]])
        gold_t, agent_t = gold_cache[iid], (patched_texts(root, [apatch]) if apatch.strip() else None)
        rec = {"run": rname, "task": iid, "group": g, "necessary": x1.get("necessary"), "v1_class": x1["class"],
               "v1_resolved": x1.get("resolved")}
        if gold_t is None or agent_t is None:
            rec.update({"class": "conflict", "detail": "patch does not apply"})
            swaps.append(rec)
            continue
        paths = {f.split("::")[0] for f in g.split(" | ")} | set(agent_t)
        base_t = {p: ((root / p).read_text(errors="replace") if (root / p).is_file() else None) for p in paths}
        full_gold = {**{p: base_t.get(p) for p in paths}, **gold_t}
        full_agent = {**{p: base_t.get(p) for p in paths}, **agent_t}
        try:
            cls_, texts, det = build_swap_v2(full_gold, full_agent, g)
        except Exception as e:  # noqa: BLE001
            cls_, texts, det = "conflict", None, f"{type(e).__name__}: {str(e)[:120]}"
        rec.update({"class": cls_, "detail": det,
                    "self_swap_identical": identity_swap(full_gold, g),
                    "agent_roundtrip_identical": identity_swap(full_agent, g)})
        if texts is not None and cls_ not in ("conflict", "identical"):
            patch = diff_from_texts(root, texts)
            k = len(swaps)
            (SWAP2_OUT / "patches" / f"{k:03d}_{rname}_{iid}.patch").write_text(patch)
            rec["patch_file"] = f"{k:03d}_{rname}_{iid}.patch"
            rec["_patch"] = patch
        swaps.append(rec)
    log("A12v classes:", json.dumps({c: sum(x["class"] == c for x in swaps) for c in CLASS_ORDER}))

    out_rows = []

    def run(label, patches):
        if not patches or s.hours_left() < 0.3:
            return {}
        t = time.time()
        r = verify_full(inp, patches, label)
        for iid, v in r.items():
            out_rows.append({"task": iid, "mode": label, **v})
        (SWAP2_OUT / "results.jsonl").write_text("".join(json.dumps(x) + "\n" for x in out_rows))
        log(f"{label}: {sum(v['resolved'] for v in r.values())}/{len(r)} resolved in {(time.time() - t) / 60:.1f} min")
        return r

    gold_ok = run("gold_rerun", {i: gold[i] for i in sorted({x["task"] for x in swaps})})
    todo = [x for x in swaps if "_patch" in x and gold_ok.get(x["task"], {}).get("resolved")]
    k = 0
    while todo:
        batch, rest = {}, []
        for x in todo:
            (rest.append(x) if x["task"] in batch else batch.__setitem__(x["task"], x))
        r = run(f"swap{k}", {t: x["_patch"] for t, x in batch.items()})
        for t, x in batch.items():
            v = r.get(t)
            if v:
                x.update({"resolved": v["resolved"], "exit": v["exit"], "tail": v["tail"], "error": v["error"]})
                x["failure_type"] = failure_type(x)
                # POST HOC (added after the smoke run, not used by the decision rule): the hybrid fails because gold's
                # other sites expect an attribute/parameter the agent's version does not provide (reverse coupling)
                x["posthoc_reverse_coupling"] = bool(re.search(
                    r"AttributeError: .* has no attribute|TypeError: .*(unexpected keyword|missing \d+ required|"
                    r"takes \d+ positional|got an unexpected)", x.get("tail") or ""))
        if not r:
            break
        todo, k = rest, k + 1
    for x in swaps:
        x["gold_ok"] = bool(gold_ok.get(x["task"], {}).get("resolved"))
        x["self_swap_pass"] = x.get("self_swap_identical") is True and x["gold_ok"]
    (SWAP2_OUT / "swaps_v2.json").write_text(json.dumps([{k_: v for k_, v in x.items() if k_ != "_patch"} for x in swaps],
                                                        indent=1))

    prim = [x for x in swaps if x["class"] == "validated-clean" and x.get("necessary") and x["self_swap_pass"]
            and x.get("resolved") is not None]

    def rates(xs, other_as_behav):
        bad = [x for x in xs if not x["resolved"] and (x["failure_type"] == "behavioural"
                                                      or (other_as_behav and x["failure_type"] == "other"))]
        by = {}
        for x in xs:
            by.setdefault(x["task"], []).append(x in bad)
        macro = sum(sum(v) / len(v) for v in by.values()) / len(by) if by else None
        rnd = _r.Random(0)
        ts = list(by)
        boots = []
        for _ in range(10000):
            smp = [rnd.choice(ts) for _ in ts] if ts else []
            vals = [b for t in smp for b in by[t]]
            boots.append(sum(vals) / len(vals) if vals else 0)
        boots.sort()
        return {"behavioural": len(bad), "n": len(xs), "rate": round(len(bad) / len(xs), 3) if xs else None,
                "task_macro": round(macro, 3) if macro is not None else None, "tasks": len(by),
                "tasks_with_behavioural_failure": sum(any(v) for v in by.values()),
                "ci95_task_bootstrap": [round(boots[250], 3), round(boots[9750], 3)] if ts else None}

    ftypes = {}
    for x in prim:
        ftypes[x.get("failure_type", "?")] = ftypes.get(x.get("failure_type", "?"), 0) + 1
    report = {
        "swaps": len(swaps),
        "classes_v2": {c: sum(x["class"] == c for x in swaps) for c in CLASS_ORDER},
        "v1_to_v2": {f"{a}->{b}": n for (a, b), n in sorted(
            {(x["v1_class"], x["class"]): sum(1 for y in swaps if (y["v1_class"], y["class"]) == (x["v1_class"], x["class"]))
             for x in swaps}.items())},
        "v1_primary_n": sum(1 for x in swaps if x["v1_class"] == "clean" and x.get("necessary")),
        "self_swap_identical": sum(bool(x.get("self_swap_identical")) for x in swaps),
        "agent_roundtrip_identical": sum(bool(x.get("agent_roundtrip_identical")) for x in swaps),
        "gold_rerun_failed": sorted(t for t, v in gold_ok.items() if not v.get("resolved")),
        "primary_failure_types": ftypes,
        "primary_other_as_behavioural": rates(prim, True),
        "primary_other_as_not": rates(prim, False),
        "POSTHOC_primary_excluding_reverse_coupling": rates([x for x in prim if not x.get("posthoc_reverse_coupling")], True),
        "POSTHOC_reverse_coupling_failures_in_primary": sum(bool(x.get("posthoc_reverse_coupling")) and not x["resolved"]
                                                            for x in prim),
    }
    # A5 v2: shared edited identifiers without keywords/builtins/self/cls, uncapped, micro and macro
    try:
        import builtins
        stop = set(keyword.kwlist) | set(dir(builtins)) | {"self", "cls"}
        micro_n = micro_hit = 0
        per = {}
        for iid, labs in nec_new.items():
            necs = [g for g, v in labs.items() if v]
            if len(necs) < 2 or iid not in groups:
                continue
            _, files, gmap, _ = groups[iid]

            def idents(g):
                toks = set()
                for fi, hi in gmap.get(g, []):
                    for line in files[fi][1][hi][2][1:]:
                        if line[:1] in "+-":
                            toks |= set(re.findall(r"[A-Za-z_][A-Za-z0-9_]{2,}", line[1:]))
                return toks - stop
            for a in necs:
                for b in necs:
                    if a != b:
                        hit = bool(idents(a) & idents(b))
                        micro_n += 1
                        micro_hit += hit
                        per.setdefault(iid, []).append(hit)
        report["A5_v2_shared_identifier"] = {"pairs": micro_n, "micro": round(micro_hit / max(1, micro_n), 3),
                                             "macro": round(sum(sum(v) / len(v) for v in per.values()) / max(1, len(per)), 3),
                                             "tasks": len(per)}
    except Exception as e:  # noqa: BLE001
        report["A5_v2_error"] = f"{type(e).__name__}: {e}"
    report["hours"] = round((time.time() - t0) / 3600, 2)
    (WORK / "siteswap2_report.json").write_text(json.dumps(report, indent=1))
    log("A12v:", json.dumps({k_: report[k_] for k_ in ("classes_v2", "primary_failure_types",
                                                        "primary_other_as_behavioural", "primary_other_as_not")}))
    log("A5v2:", json.dumps(report.get("A5_v2_shared_identifier") or report.get("A5_v2_error")))
````

### `evalkit/regrade.py`
R1 stray-file re-grade (CPU, offline, no model). Pre-registered in docs/experiments.md (Oct 28).

````python
%%writefile /kaggle/working/evalkit/regrade.py
"""
R1 stray-file re-grade (CPU, offline, no model). Pre-registered in docs/experiments.md (Oct 28).
Question: do stray scratch files (repro.py, reproduce_issue.py, non-.py files; trace_analyzer's `stray` rule) make
otherwise-correct patches fail, or are they only a symptom of runs that ran out of budget?
For every valid-task run in the embedded payload (same runs as A3/A4) whose patch has stray files:
  orig      the agent's patch as submitted (control: re-grade noise)
  stripped  the same patch minus the stray files' diff blocks
Resolved cases are included too (stripping must not break them). Output: /kaggle/working/regrade/report.json.
Run inside the venv: python -m evalkit.regrade
"""
import json
import time

from .compat import patch_sandbox_venv, use_scorer_sandbox
from .config import VALID_TASKS, WORK
from .inputs import Inputs
from .necessity import path_of, runs_from_payload, split_patch, verify_full
from .server import make_sandbox_offline, set_scorer_env
from .util import log

OUT = WORK / "regrade"


def strip_files(patch, drop):
    """The patch without the diff blocks of the files in `drop`."""
    keep = []
    for header, hunks in split_patch(patch):
        if path_of(header) in drop or any(h.startswith("+++ b/") and h[6:].strip() in drop for h in header):
            continue
        keep.append("".join(header) + "".join("".join(h[2]) for h in hunks))
    return "".join(keep)


def main():
    t0 = time.time()
    set_scorer_env()
    patch_sandbox_venv()
    inp = Inputs.locate(False, need_model=False)
    use_scorer_sandbox(inp.data)
    make_sandbox_offline()
    OUT.mkdir(parents=True, exist_ok=True)
    valid = set(json.loads(VALID_TASKS.read_text()))
    runs, _ = runs_from_payload()
    cases = []
    for run, rows in sorted(runs.items()):
        for iid, (row, patch) in sorted(rows.items()):
            stray = row.get("stray_files_in_patch") or []
            if iid not in valid or not stray or not patch.strip():
                continue
            stripped = strip_files(patch, set(stray))
            cases.append({"run": run, "id": iid, "resolved_in_run": bool(row["resolved"]), "stray": stray,
                          "orig": patch, "stripped": stripped, "stripped_empty": not stripped.strip()})
    log(f"R1: {len(cases)} stray-file cases ({sum(not c['resolved_in_run'] for c in cases)} unresolved in run)")
    results = []
    for mode in ("orig", "stripped"):  # one Evaluator call per (mode, run): task ids must be unique per call
        for run in sorted({c["run"] for c in cases}):
            batch = [c for c in cases if c["run"] == run and not (mode == "stripped" and c["stripped_empty"])]
            if not batch:
                continue
            res = verify_full(inp, {c["id"]: c[mode] for c in batch}, f"r1_{mode}_{run}")
            for c in batch:
                r = res.get(c["id"], {})
                results.append({"run": run, "id": c["id"], "mode": mode, "resolved_in_run": c["resolved_in_run"],
                                "stray": c["stray"], "resolved": r.get("resolved"), "exit": r.get("exit"),
                                "error": r.get("error"), "tail": (r.get("tail") or "")[-1500:]})
            log(f"  {mode} {run}: {sum(bool(results[-k]['resolved']) for k in range(1, len(batch) + 1))}/{len(batch)}")
    (OUT / "results.jsonl").write_text("".join(json.dumps(r) + "\n" for r in results))
    (OUT / "patches.json").write_text(json.dumps(cases, indent=1))

    def tally(unresolved_in_run, mode):
        rs = [r for r in results if r["mode"] == mode and r["resolved_in_run"] is not unresolved_in_run]
        return {"n": len(rs), "resolved": sum(bool(r["resolved"]) for r in rs)}
    key = lambda r: (r["run"], r["id"])
    orig = {key(r): r["resolved"] for r in results if r["mode"] == "orig"}
    rescued = [r for r in results if r["mode"] == "stripped" and r["resolved"] and not orig.get(key(r))]
    broken = [r for r in results if r["mode"] == "stripped" and not r["resolved"] and orig.get(key(r))]
    report = {"cases": len(cases), "stripped_empty": sum(c["stripped_empty"] for c in cases),
              "unresolved_in_run": {"orig": tally(True, "orig"), "stripped": tally(True, "stripped")},
              "resolved_in_run": {"orig": tally(False, "orig"), "stripped": tally(False, "stripped")},
              "rescued_by_stripping": [key(r) for r in rescued], "broken_by_stripping": [key(r) for r in broken],
              "orig_disagrees_with_run": [k for k, v in orig.items()
                                          if bool(v) != next(c["resolved_in_run"] for c in cases if (c["run"], c["id"]) == k)],
              "minutes": round((time.time() - t0) / 60, 1)}
    (WORK / "regrade_report.json").write_text(json.dumps(report, indent=1))
    log("R1 REPORT", json.dumps(report))


if __name__ == "__main__":
    main()
````

### `evalkit/factory_smoke.py`
FS1 factory rollout smoke (GPU). Pre-registered in docs/experiments.md (Oct 28).

````python
%%writefile /kaggle/working/evalkit/factory_smoke.py
"""
FS1 factory rollout smoke (GPU). Pre-registered in docs/experiments.md (Oct 28).
60 of the verified factory-v0 tasks (dataset gemma4-factory-v0; 20 per library, stratified, seed 0):
  1. rewrite   Gemma (thinking off) turns each template issue into a user-style bug report, using the hidden tests'
               inputs/expectations as the reproducer. A rewrite that quotes a gold-patch added line, names a test, or
               is too short falls back to the template (counted).
  2. rollouts  the submission under test (v14) unchanged: the first 6 tasks at concurrency 1 (throughput baseline +
               health check), the other 54 at concurrency 6.
Output: /kaggle/working/factory_smoke_report.json + eval_smoke_c{1,6}.tgz (traces) + smoke_tasks.jsonl.
Run inside the venv: python -m evalkit.factory_smoke
"""
import json
import random
import re
import time
import urllib.request
from concurrent.futures import ThreadPoolExecutor
from dataclasses import replace

import litellm

from .compat import patch_sandbox_venv, use_scorer_sandbox
from .config import INPUT, WORK, Settings
from .inputs import Inputs
from .report import Report
from .runner import ChunkedEvaluation
from .server import ScorerServer, make_sandbox_offline, set_scorer_env
from .submission import Submission
from .util import log

PER_LIB, N_BASE, CONC = 20, 6, 6
TASKS_OUT = WORK / "smoke_tasks.jsonl"
REWRITE_PROMPT = """You are a user of the Python library `{pkg}` filing a GitHub issue about a bug you hit.

Below are (1) a terse draft of the issue and (2) the maintainers' internal test code that exposes the bug.
Write the GitHub issue as a real user would: a short title line, what you were doing, a minimal standalone code
snippet that calls the public API (derive the inputs and expected values from the test code), what you got
(exception or wrong output) and what you expected.

Rules:
- Do NOT mention tests, test names, test classes, pytest, or any file under tests/.
- Do NOT say where in the library's source the bug is, and do NOT suggest a fix.
- Plain markdown, under 250 words. Output only the issue text.

(1) Draft issue:
{issue}

(2) Test code:
```python
{tests}
```"""


def lib_of(iid):
    return iid.split("_", 1)[1].rsplit("_", 1)[0]


def pick(tasks, meta):
    """PER_LIB tasks per library, singles and pairs in their natural proportion (shuffled with seed 0)."""
    rng = random.Random(0)
    out = []
    for lib in sorted({lib_of(t["instance_id"]) for t in tasks}):
        pool = [t for t in tasks if lib_of(t["instance_id"]) == lib]
        rng.shuffle(pool)
        out += pool[:PER_LIB]
    rng.shuffle(out)
    return out


def added_lines(patch):
    return [l[1:].strip() for l in patch.splitlines() if l.startswith("+") and not l.startswith("+++")]


def check_rewrite(text, task):
    if len(text) < 200:
        return "short"
    if re.search(r"\btests?/|def test_|pytest|\bTestCase\b", text):
        return "names_tests"
    f2p = task["FAIL_TO_PASS"] if isinstance(task["FAIL_TO_PASS"], list) else json.loads(task["FAIL_TO_PASS"])
    if any(n.rsplit("::", 1)[-1] in text for n in f2p):
        return "names_tests"
    if any(len(l) >= 8 and l in text for l in added_lines(task["patch"])):
        return "leaks_gold"
    return None


def chat(base_url, model, prompt):
    body = {"model": model, "messages": [{"role": "user", "content": prompt}], "temperature": 0.7, "top_p": 0.95,
            "max_tokens": 1024, "chat_template_kwargs": {"enable_thinking": False}}
    req = urllib.request.Request(f"{base_url}/chat/completions", data=json.dumps(body).encode(),
                                 headers={"Content-Type": "application/json"})
    out = json.load(urllib.request.urlopen(req, timeout=600))
    return (out["choices"][0]["message"].get("content") or "").strip()


def rewrite_all(server, tasks):
    base_url = server.server.base_url
    model = json.load(urllib.request.urlopen(f"{base_url}/models"))["data"][0]["id"]

    def one(t):
        tests = "\n".join(l[1:] for l in t["test_patch"].splitlines() if l.startswith("+") and not l.startswith("+++"))
        prompt = REWRITE_PROMPT.format(pkg=t["repo"].split("/")[-1], issue=t["problem_statement"], tests=tests[:3000])
        for _ in range(2):  # one retry on a rejected rewrite
            try:
                text = chat(base_url, model, prompt)
            except Exception as e:  # noqa: BLE001
                text, why = "", f"error: {type(e).__name__}"
            else:
                why = check_rewrite(text, t)
            if why is None:
                return text, None
        return None, why

    t0 = time.time()
    with ThreadPoolExecutor(16) as ex:
        outs = list(ex.map(one, tasks))
    stats = {"n": len(tasks), "rewritten": 0, "fallback": {}, "minutes": 0}
    stats["templates"] = {t["instance_id"]: t["problem_statement"] for t in tasks}
    for t, (text, why) in zip(tasks, outs):
        if text:
            t["problem_statement"] = text
            stats["rewritten"] += 1
        else:
            stats["fallback"][why] = stats["fallback"].get(why, 0) + 1
    stats["minutes"] = round((time.time() - t0) / 60, 1)
    stats["samples"] = {t["instance_id"]: t["problem_statement"] for t in tasks[:3]}
    log("REWRITE", json.dumps({k: v for k, v in stats.items() if k != "samples"}))
    return stats


def phase(s, inp, sub, models, adapters, label, ids, conc):
    ps = replace(s, name=label, task_ids=ids, concurrency=conc, chunk_size=len(ids) if conc == 1 else 18,
                 max_tasks=None, repeats=1)
    t = time.time()
    ChunkedEvaluation(ps, inp, sub, models, adapters).run()
    wall = time.time() - t
    report = Report(ps, inp)
    summary = report.build()
    p = report.root / "rep0" / "merged" / "task_results.jsonl"
    rows = [json.loads(l) for l in open(p) if l.strip()] if p.exists() else []
    # not report.package(): it deletes submission_under_test, which the next phase still needs
    log.sh(f"cd {WORK} && tar -czf {ps.label}.tgz {ps.label} && rm -rf {ps.label}")
    return {"tasks": len(ids), "done": len(rows), "resolved": sum(bool(r.get("resolved")) for r in rows),
            "wall_min": round(wall / 60, 1), "tasks_per_h": round(len(rows) / (wall / 3600), 2) if rows else 0,
            "task_sec_median": sorted(r.get("duration_seconds") or 0 for r in rows)[len(rows) // 2] if rows else None,
            "summary": summary}, rows


def breakdown(rows, meta, budget):
    by = {}
    cap = (budget.get("max_time_minutes") or 8) * 60
    for r in rows:
        m = meta.get(r["instance_id"], {})
        for key in (f"lib:{lib_of(r['instance_id'])}", f"kind:{m.get('kind')}", *(f"op:{o}" for o in m.get("ops", []))):
            b = by.setdefault(key, [0, 0])
            b[0] += bool(r.get("resolved"))
            b[1] += 1
    return {"by": {k: f"{a}/{n}" for k, (a, n) in sorted(by.items())},
            "time_cap_hits": sum((r.get("duration_seconds") or 0) >= cap - 5 for r in rows),
            "call_cap_hits": sum((r.get("tool_calls") or 0) >= (budget.get("max_tool_calls") or 28) - 1 for r in rows),
            "empty_patch": sum(not r.get("agent_patch_size") for r in rows),
            "errors": sum(bool(r.get("error")) for r in rows)}


def main():
    s = Settings.load()
    set_scorer_env()
    patch_sandbox_venv()
    litellm.drop_params = True
    comp = Inputs.locate(False)
    use_scorer_sandbox(comp.data)
    fac = sorted(p for p in INPUT.rglob("tasks.jsonl") if (p.parent / "task_meta.jsonl").exists())
    assert fac, "attach dataset gemma4-factory-v0"
    fdir = fac[0].parent
    meta = {json.loads(l)["id"]: json.loads(l) for l in open(fdir / "task_meta.jsonl")}
    tasks = pick([json.loads(l) for l in open(fac[0]) if l.strip()], meta)
    log(f"factory: {fdir}, picked {len(tasks)} tasks")

    sub = Submission().unpack()
    declared, adapters = sub.declared_model(), sub.adapters()
    server = ScorerServer(comp.model_path, adapters).start()
    models = server.registry(declared)
    report = {"tasks": [t["instance_id"] for t in tasks]}
    try:
        report["rewrite"] = rewrite_all(server, tasks)
        with open(TASKS_OUT, "w") as f:  # same keys as the competition rows; templates are in the report
            f.writelines(json.dumps(t) + "\n" for t in tasks)
        make_sandbox_offline()
        inp = Inputs(TASKS_OUT, comp.data, fdir / "snapshots", comp.model_path)
        ids = [t["instance_id"] for t in tasks]
        report["c1"], rows1 = phase(s, inp, sub, models, adapters, "smoke_c1", ids[:N_BASE], 1)
        log("PHASE c1", json.dumps({k: v for k, v in report["c1"].items() if k != "summary"}))
        if report["c1"]["done"] and s.hours_left() > 0.5:
            report["c6"], rows6 = phase(s, inp, sub, models, adapters, "smoke_c6", ids[N_BASE:], CONC)
            log("PHASE c6", json.dumps({k: v for k, v in report["c6"].items() if k != "summary"}))
        else:
            rows6 = []
        rows = rows1 + rows6
        report["overall"] = {"resolved": sum(bool(r.get("resolved")) for r in rows), "n": len(rows),
                             **breakdown(rows, meta, sub.eval_budget())}
        if report.get("c6") and report["c1"]["tasks_per_h"]:
            report["speedup_c6_vs_c1"] = round(report["c6"]["tasks_per_h"] / report["c1"]["tasks_per_h"], 2)
    finally:
        (WORK / "factory_smoke_report.json").write_text(json.dumps(report, indent=1))
        log("FS1 REPORT", json.dumps({k: v for k, v in report.items() if k not in ("tasks", "rewrite")}))
        log.sh(f"cd {WORK} && tar -czf factory_smoke.tgz factory_smoke_report.json smoke_tasks.jsonl run.log "
               f"$(ls eval_smoke_*.tgz 2>/dev/null)")
        server.stop()
        log(f"RUN_DONE in {(time.time() - s.t0) / 3600:.1f} h")


if __name__ == "__main__":
    main()
````

### `evalkit/main.py`
Everything after the env step. Runs inside the scorer venv: python -m evalkit.main (cwd /kaggle/working).

````python
%%writefile /kaggle/working/evalkit/main.py
"""Everything after the env step. Runs inside the scorer venv: python -m evalkit.main (cwd /kaggle/working)."""
import time

import litellm

from .compat import patch_sandbox_venv, use_scorer_sandbox
from .config import Settings
from .inputs import Inputs
from .report import Report
from .runner import ChunkedEvaluation
from .server import ScorerServer, make_sandbox_offline, set_scorer_env
from .submission import Submission
from .util import log


def main():
    s = Settings.load()
    set_scorer_env()
    patch_sandbox_venv()
    litellm.drop_params = True
    inp = Inputs.locate(s.devset)
    use_scorer_sandbox(inp.data)
    sub = Submission().unpack()
    declared, adapters = sub.declared_model(), sub.adapters()
    log("declared model:", declared, "| adapters:", list(getattr(adapters, "adapters", {}) or {}))
    server = ScorerServer(inp.model_path, adapters).start()
    models = server.registry(declared)
    make_sandbox_offline()
    report = Report(s, inp)
    try:
        ChunkedEvaluation(s, inp, sub, models, adapters).run()
    finally:
        report.build()
        report.package()
        server.stop()
        log(f"RUN_DONE in {(time.time() - s.t0) / 3600:.1f} h")


if __name__ == "__main__":
    main()
````
